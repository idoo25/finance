```python
# @title Imports & Configuration
import yfinance as yf
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
import gradio as gr
import requests
import io
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# ==================== CONFIGURATION ====================

# Israeli CPI Inflation Data (Source: CBS Israel)
INFLATION_DATA = {
    2004: 1.2, 2005: 2.4, 2006: -0.1, 2007: 3.4, 2008: 3.8, 2009: 3.9,
    2010: 2.7, 2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
    2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8, 2019: 0.6,
    2020: -0.7, 2021: 2.8, 2022: 5.3, 2023: 3.0, 2024: 3.2,
    2025: 2.4
}

# Portfolio Definitions
PORTFOLIOS = {
    'Golden Butterfly': ['VTI', 'VBR', 'TLT', 'SHY', 'GLD'],
    'Ivy Portfolio': ['VTI', 'VEU', 'AGG', 'DBC', 'VNQ'],
    'S&P 500': ['IVV'],
    'Nasdaq 100': ['QQQ'],
    'Semiconductors': ['SOXX'],
    'Gold': ['IAU'],
    'Silver': ['SLV']
}

USD_ILS_TICKER = 'ILS=X'
TRADING_DAYS = 252

# Color scheme
COLORS = {
    'Golden Butterfly': '#D4AF37',
    'Ivy Portfolio': '#228B22',
    'S&P 500': '#1E90FF',
    'Nasdaq 100': '#000000',
    'Semiconductors': '#8B008B',
    'Gold': '#FFD700',
    'Silver': '#C0C0C0'
}

print("✅ Configuration loaded!")
```

```python
# @title Core Analysis Functions

def build_cpi_index(date_index, base_value=100.0):
    """Build daily CPI index from annual inflation rates."""
    cpi_daily = pd.Series(index=date_index, dtype=float)
    cpi_daily.iloc[0] = base_value
    current_cpi = base_value
    
    for year in sorted(INFLATION_DATA.keys()):
        if year >= date_index[0].year:
            rate = INFLATION_DATA[year] / 100
            current_cpi *= (1 + rate)
            year_days = date_index[date_index.year == year]
            if len(year_days) > 0:
                cpi_daily.loc[year_days[-1]] = current_cpi
    
    return cpi_daily.interpolate(method='time').ffill()


# ============ 1. SINGLE ASSET ANALYSIS ============
def analyze_single_asset(asset_name, period):
    """Analyze a single asset USD vs ILS."""
    tickers = PORTFOLIOS.get(asset_name, [asset_name])
    ticker = tickers[0]
    
    data = yf.download([ticker, USD_ILS_TICKER], period=period, progress=False)['Close']
    data = data.ffill().dropna()
    
    if data.empty:
        return None, "No data available"
    
    asset_usd = data[ticker].to_numpy()
    usd_ils = data[USD_ILS_TICKER].to_numpy()
    asset_ils = np.multiply(asset_usd, usd_ils)
    
    log_ret_usd = np.diff(np.log(asset_usd))
    log_ret_ils = np.diff(np.log(asset_ils))
    
    cum_usd = np.insert(np.exp(np.cumsum(log_ret_usd)) - 1, 0, 0)
    cum_ils = np.insert(np.exp(np.cumsum(log_ret_ils)) - 1, 0, 0)
    
    vol_usd = np.std(log_ret_usd) * np.sqrt(TRADING_DAYS) * 100
    vol_ils = np.std(log_ret_ils) * np.sqrt(TRADING_DAYS) * 100
    correlation = np.corrcoef(log_ret_usd, log_ret_ils)[0, 1]
    
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=data.index, y=cum_usd * 100, name=f'{asset_name} (USD)',
                             line=dict(color='#1E90FF', width=2)))
    fig.add_trace(go.Scatter(x=data.index, y=cum_ils * 100, name=f'{asset_name} (ILS)',
                             line=dict(color='#228B22', width=2, dash='dash')))
    
    fig.update_layout(
        title=f'{asset_name}: USD vs ILS Returns ({period})',
        xaxis_title='Date', yaxis_title='Cumulative Return (%)',
        hovermode='x unified', template='plotly_white',
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
    )
    
    stats = f"""### 📊 {asset_name} Analysis Results\n\n| Metric | USD | ILS |\n|--------|-----|-----|\n| **Return** | {cum_usd[-1]*100:.2f}% | {cum_ils[-1]*100:.2f}% |\n| **Volatility** | {vol_usd:.2f}% | {vol_ils:.2f}% |\n\n**Currency Impact:** {(cum_ils[-1] - cum_usd[-1])*100:+.2f}%\n**Correlation:** {correlation:.4f}"""
    
    return fig, stats


# ============ 2. S&P 500 SYNTHETIC IL FUND ============
def analyze_sp500_synthetic_il(period):
    """Simulate Israeli S&P 500 fund (unhedged) - like Tachlit/Kesem funds."""
    tickers = ['IVV', USD_ILS_TICKER]
    
    data = yf.download(tickers, period=period, progress=False)['Close']
    data = data.ffill().dropna()
    
    if data.empty:
        return None, "No data available"
    
    sp500_usd = data['IVV'].to_numpy()
    usd_ils = data[USD_ILS_TICKER].to_numpy()
    sp500_ils_synthetic = np.multiply(sp500_usd, usd_ils)
    
    log_ret_usd = np.diff(np.log(sp500_usd))
    log_ret_ils = np.diff(np.log(sp500_ils_synthetic))
    
    cum_ret_usd = np.insert(np.exp(np.cumsum(log_ret_usd)) - 1, 0, 0)
    cum_ret_ils = np.insert(np.exp(np.cumsum(log_ret_ils)) - 1, 0, 0)
    
    vol_usd = np.std(log_ret_usd) * np.sqrt(TRADING_DAYS) * 100
    vol_ils = np.std(log_ret_ils) * np.sqrt(TRADING_DAYS) * 100
    gap = (cum_ret_ils[-1] - cum_ret_usd[-1]) * 100
    
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=data.index, y=cum_ret_usd * 100, name='S&P 500 (USD)',
                             line=dict(color='#1E90FF', width=2)))
    fig.add_trace(go.Scatter(x=data.index, y=cum_ret_ils * 100, name='Israeli Fund Simulation (ILS)',
                             line=dict(color='#228B22', width=2, dash='dash')))
    
    fig.update_layout(
        title=f'S&P 500: USD vs Israeli Fund Simulation ({period})',
        xaxis_title='Date', yaxis_title='Cumulative Return (%)',
        hovermode='x unified', template='plotly_white'
    )
    
    stats = f"""### 🏦 S&P 500 Israeli Fund Simulation\n\nThis simulates an **unhedged Israeli S&P 500 fund** (like Tachlit/Kesem).\n\n| Metric | USD (Direct) | ILS (Israeli Fund) |\n|--------|--------------|-------------------|\n| **Cumulative Return** | {cum_ret_usd[-1]*100:.2f}% | {cum_ret_ils[-1]*100:.2f}% |\n| **Volatility (Annual)** | {vol_usd:.2f}% | {vol_ils:.2f}% |\n\n**Currency Impact (USD/ILS):** {gap:+.2f}%\n\n💡 *Positive = ILS weakened (good for Israeli investors)*"""
    
    return fig, stats


# ============ 3. TASE API (Tel Aviv Stock Exchange) ============
def get_tase_history(security_id, years):
    """Fetch historical data from Tel Aviv Stock Exchange API."""
    url = "https://api.tase.co.il/api/Chart/History"
    today = datetime.now()
    start_date = today - timedelta(days=int(years) * 365)
    
    params = {"pType": 1, "id": security_id, "l": 4, "d": 0}
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "Referer": "https://www.tase.co.il/",
        "Accept-Language": "he-IL,he;q=0.9,en-US;q=0.8,en;q=0.7"
    }
    
    try:
        response = requests.get(url, params=params, headers=headers, timeout=30)
        response.raise_for_status()
        data = response.json()
        
        if 'c' not in data:
            return None, "No data in API response"
        
        df = pd.DataFrame(data['c'])
        df = df[['d', 'c']]
        df.columns = ['Date', 'Close']
        
        df['Date'] = df['Date'].astype(str).str.extract(r'(\d+)').astype(int)
        df['Date'] = pd.to_datetime(df['Date'], unit='ms')
        df.set_index('Date', inplace=True)
        df = df[df.index >= start_date]
        
        if df.empty:
            return None, "No data for selected period"
        
        # Calculate returns
        df['Return'] = (df['Close'] / df['Close'].iloc[0] - 1) * 100
        
        fig = go.Figure()
        fig.add_trace(go.Scatter(x=df.index, y=df['Return'], name=f'Security {security_id}',
                                 line=dict(color='#1E90FF', width=2)))
        
        fig.update_layout(
            title=f'TASE Security {security_id} ({years} Years)',
            xaxis_title='Date', yaxis_title='Return (%)',
            hovermode='x unified', template='plotly_white'
        )
        
        total_return = df['Return'].iloc[-1]
        stats = f"""### 🇮🇱 TASE Security {security_id}\n\n| Metric | Value |\n|--------|-------|\n| **Data Points** | {len(df)} |\n| **Start Date** | {df.index[0].strftime('%Y-%m-%d')} |\n| **End Date** | {df.index[-1].strftime('%Y-%m-%d')} |\n| **Total Return** | {total_return:.2f}% |\n| **Start Price** | ₪{df['Close'].iloc[0]:.2f} |\n| **End Price** | ₪{df['Close'].iloc[-1]:.2f} |"""
        
        return fig, stats
        
    except Exception as e:
        return None, f"Error fetching TASE data: {str(e)}"


# ============ 4. CSV FILE UPLOAD ANALYSIS ============
def analyze_uploaded_csv(file, filter_last_year):
    """Analyze uploaded CSV file with fund data and compare with S&P 500."""
    if file is None:
        return None, "Please upload a CSV file"
    
    try:
        df = pd.read_csv(file.name)
    except Exception as e:
        return None, f"Error reading file: {str(e)}"
    
    # Detect column names (Hebrew or English)
    if 'תאריך' in df.columns:
        date_col, price_col = 'תאריך', 'מחיר פדיון'
    elif 'Date' in df.columns:
        date_col, price_col = 'Date', 'Price'
    else:
        # Try to guess by position
        date_col = df.columns[1] if len(df.columns) > 1 else df.columns[0]
        price_col = df.columns[3] if len(df.columns) > 3 else df.columns[-1]
    
    df = df.rename(columns={date_col: 'Date', price_col: 'IL_Price'})
    df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
    df = df.dropna(subset=['Date']).sort_values('Date').set_index('Date')
    
    # Clean price column
    if df['IL_Price'].dtype == object:
        df['IL_Price'] = df['IL_Price'].astype(str).str.replace(',', '').astype(float)
    
    # Filter to last year if requested
    if filter_last_year:
        last_date = df.index[-1]
        start_date = last_date - timedelta(days=365)
        df = df[df.index >= start_date]
    
    if df.empty:
        return None, "No valid data in file"
    
    # Calculate returns
    df['IL_Return'] = (df['IL_Price'] / df['IL_Price'].iloc[0] - 1) * 100
    
    # Download S&P 500 for comparison
    start_date = df.index[0]
    end_date = df.index[-1] + timedelta(days=5)
    
    us_data = yf.download('IVV', start=start_date, end=end_date, progress=False)['Close']
    
    # Merge datasets
    combined = pd.merge_asof(
        df[['IL_Return']], 
        us_data.to_frame('US_Price'),
        left_index=True, right_index=True, direction='nearest'
    )
    combined['US_Return'] = (combined['US_Price'] / combined['US_Price'].iloc[0] - 1) * 100
    
    # Create chart
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=combined.index, y=combined['IL_Return'], 
                             name='Your Fund (ILS)', line=dict(color='#228B22', width=2)))
    fig.add_trace(go.Scatter(x=combined.index, y=combined['US_Return'], 
                             name='S&P 500 (USD)', line=dict(color='#1E90FF', width=2, dash='dash')))
    
    fig.update_layout(
        title='Your Fund vs S&P 500',
        xaxis_title='Date', yaxis_title='Return (%)',
        hovermode='x unified', template='plotly_white'
    )
    
    il_final = combined['IL_Return'].iloc[-1]
    us_final = combined['US_Return'].iloc[-1]
    diff = il_final - us_final
    
    stats = f"""### 📁 Uploaded Fund Analysis\n\n| Metric | Your Fund | S&P 500 |\n|--------|-----------|---------|\n| **Period** | {df.index[0].strftime('%Y-%m-%d')} to {df.index[-1].strftime('%Y-%m-%d')} | Same |\n| **Total Return** | {il_final:.2f}% | {us_final:.2f}% |\n| **Data Points** | {len(df)} | {len(combined)} |\n\n**Difference vs S&P 500:** {diff:+.2f}%\n\n{'✅ Outperformed S&P 500!' if diff > 0 else '❌ Underperformed S&P 500'}"""
    
    return fig, stats


# ============ 5. PORTFOLIO COMPARISON ============
def compare_portfolios_analysis(selected_portfolios, start_year, show_inflation):
    """Compare multiple portfolios."""
    if not selected_portfolios:
        return None, "Please select at least one portfolio"
    
    all_tickers = [USD_ILS_TICKER]
    for name in selected_portfolios:
        all_tickers.extend(PORTFOLIOS[name])
    all_tickers = list(set(all_tickers))
    
    data = yf.download(all_tickers, start=f"{int(start_year)}-01-01", progress=False)['Close']
    data = data.ffill().dropna()
    
    if data.empty:
        return None, "No data available"
    
    usd_ils = data[USD_ILS_TICKER]
    cpi = build_cpi_index(data.index) if show_inflation else None
    
    fig = make_subplots(rows=1, cols=2 if show_inflation else 1,
                        subplot_titles=['Nominal Returns (ILS)', 'Real Returns (Inflation-Adjusted)'] if show_inflation else ['Nominal Returns (ILS)'])
    
    results = {}
    
    for name in selected_portfolios:
        tickers = PORTFOLIOS[name]
        color = COLORS.get(name, '#888888')
        
        if len(tickers) == 1:
            prices_ils = data[tickers[0]] * usd_ils
            nom_returns = (prices_ils / prices_ils.iloc[0]) * 100
        else:
            basket_ils = data[tickers].multiply(usd_ils, axis=0)
            daily_rets = basket_ils.pct_change().mean(axis=1)
            nom_returns = (1 + daily_rets).cumprod() * 100
        
        fig.add_trace(go.Scatter(x=data.index, y=nom_returns, name=name,
                                 line=dict(color=color, width=2)), row=1, col=1)
        
        if show_inflation:
            real_returns = nom_returns / cpi * 100
            fig.add_trace(go.Scatter(x=data.index, y=real_returns, name=f'{name}',
                                     line=dict(color=color, width=2), showlegend=False), row=1, col=2)
            results[name] = {'nominal': nom_returns.iloc[-1] - 100, 'real': real_returns.iloc[-1] - 100}
        else:
            results[name] = {'nominal': nom_returns.iloc[-1] - 100}
    
    if show_inflation:
        fig.add_trace(go.Scatter(x=data.index, y=cpi, name='CPI (Inflation)',
                                 line=dict(color='red', width=2, dash='dot')), row=1, col=1)
    
    fig.add_hline(y=100, line_dash="dash", line_color="gray", opacity=0.5)
    fig.update_layout(title=f'Portfolio Comparison ({int(start_year)} - Present)',
                      hovermode='x unified', template='plotly_white', height=500)
    fig.update_yaxes(title_text='Value (Base=100)')
    
    stats = f"### 📈 Portfolio Comparison Results ({int(start_year)} - Present)\n\n"
    if show_inflation:
        stats += "| Portfolio | Nominal Return | Real Return |\n|-----------|----------------|-------------|\n"
        for name, vals in results.items():
            stats += f"| **{name}** | {vals['nominal']:.2f}% | {vals['real']:.2f}% |\n"
        stats += f"\n**Total Inflation:** {cpi.iloc[-1] - 100:.2f}%"
    else:
        stats += "| Portfolio | Return |\n|-----------|--------|\n"
        for name, vals in results.items():
            stats += f"| **{name}** | {vals['nominal']:.2f}% |\n"
    
    return fig, stats


# ============ 6. YEARLY BREAKDOWN ============
def yearly_breakdown_analysis(selected_portfolios, start_year):
    """Generate yearly returns breakdown."""
    if not selected_portfolios:
        return None, pd.DataFrame()
    
    all_tickers = [USD_ILS_TICKER]
    for name in selected_portfolios:
        all_tickers.extend(PORTFOLIOS[name])
    all_tickers = list(set(all_tickers))
    
    data = yf.download(all_tickers, start=f"{int(start_year)}-01-01", progress=False)['Close']
    data = data.ffill().dropna()
    
    usd_ils = data[USD_ILS_TICKER]
    cpi = build_cpi_index(data.index)
    
    portfolio_values = pd.DataFrame(index=data.index)
    
    for name in selected_portfolios:
        tickers = PORTFOLIOS[name]
        if len(tickers) == 1:
            portfolio_values[name] = data[tickers[0]] * usd_ils
        else:
            basket_ils = data[tickers].multiply(usd_ils, axis=0)
            daily_rets = basket_ils.pct_change().mean(axis=1)
            portfolio_values[name] = (1 + daily_rets).cumprod() * 100
    
    yearly = portfolio_values.resample('YE').last()
    yearly_cpi = cpi.resample('YE').last()
    yearly_returns = yearly.pct_change() * 100
    yearly_inflation = yearly_cpi.pct_change() * 100
    
    df_plot = yearly_returns.dropna().copy()
    df_plot['Inflation'] = yearly_inflation
    df_plot.index = df_plot.index.year
    
    fig = px.imshow(df_plot.T, labels=dict(x="Year", y="Portfolio", color="Return %"),
                    color_continuous_scale='RdYlGn', color_continuous_midpoint=0, aspect='auto')
    fig.update_layout(title='Yearly Returns Heatmap', template='plotly_white', height=400)
    
    table_df = df_plot.round(2).reset_index()
    table_df.columns = ['Year'] + list(table_df.columns[1:])
    
    return fig, table_df


# ============ 7. DCA SIMULATION ============
def dca_simulation(initial_investment, monthly_investment, start_year, selected_portfolios):
    """Simulate Dollar Cost Averaging."""
    if not selected_portfolios:
        return None, pd.DataFrame()
    
    all_tickers = [USD_ILS_TICKER]
    for name in selected_portfolios:
        all_tickers.extend(PORTFOLIOS[name])
    all_tickers = list(set(all_tickers))
    
    data = yf.download(all_tickers, start=f"{int(start_year)}-01-01", progress=False)['Close']
    data = data.ffill().dropna()
    
    usd_ils = data[USD_ILS_TICKER]
    cpi = build_cpi_index(data.index)
    
    prices = pd.DataFrame(index=data.index)
    for name in selected_portfolios:
        tickers = PORTFOLIOS[name]
        if len(tickers) == 1:
            prices[name] = data[tickers[0]] * usd_ils
        else:
            basket = data[tickers].multiply(usd_ils, axis=0)
            rets = basket.pct_change().mean(axis=1)
            price_series = (1 + rets).cumprod() * 100
            price_series.iloc[0] = 100
            prices[name] = price_series
    
    monthly_prices = prices.resample('ME').last()
    monthly_cpi = cpi.reindex(monthly_prices.index, method='ffill')
    
    results = []
    portfolio_values_over_time = pd.DataFrame(index=monthly_prices.index)
    
    for name in selected_portfolios:
        units = initial_investment / monthly_prices[name].iloc[0]
        total_invested = initial_investment
        values = [units * monthly_prices[name].iloc[0]]
        
        for i in range(1, len(monthly_prices)):
            adj_contribution = monthly_investment * (monthly_cpi.iloc[i] / monthly_cpi.iloc[0])
            units += adj_contribution / monthly_prices[name].iloc[i]
            total_invested += adj_contribution
            values.append(units * monthly_prices[name].iloc[i])
        
        final_value = values[-1]
        portfolio_values_over_time[name] = values
        
        results.append({
            'Portfolio': name,
            'Total Invested': f"₪{total_invested:,.0f}",
            'Final Value': f"₪{final_value:,.0f}",
            'Return': f"{(final_value / total_invested - 1) * 100:.2f}%"
        })
    
    fig = go.Figure()
    for name in selected_portfolios:
        fig.add_trace(go.Scatter(x=monthly_prices.index, y=portfolio_values_over_time[name],
                                 name=name, line=dict(color=COLORS.get(name, '#888888'), width=2)))
    
    fig.update_layout(title=f'DCA Portfolio Growth ({int(start_year)} - Present)',
                      xaxis_title='Date', yaxis_title='Portfolio Value (₪)',
                      hovermode='x unified', template='plotly_white')
    
    return fig, pd.DataFrame(results)


# ============ 8. LUMP SUM VS DCA ============
def lump_sum_vs_dca(lump_sum, dca_total, dca_months, start_year, selected_portfolios):
    """Compare lump sum vs DCA strategies."""
    if not selected_portfolios:
        return None, pd.DataFrame()
    
    all_tickers = [USD_ILS_TICKER]
    for name in selected_portfolios:
        all_tickers.extend(PORTFOLIOS[name])
    all_tickers = list(set(all_tickers))
    
    data = yf.download(all_tickers, start=f"{int(start_year)}-01-01", progress=False)['Close']
    data = data.ffill().dropna()
    
    usd_ils = data[USD_ILS_TICKER]
    
    prices = pd.DataFrame(index=data.index)
    for name in selected_portfolios:
        tickers = PORTFOLIOS[name]
        if len(tickers) == 1:
            prices[name] = data[tickers[0]] * usd_ils
        else:
            basket = data[tickers].multiply(usd_ils, axis=0)
            rets = basket.pct_change().mean(axis=1)
            p = (1 + rets).cumprod() * 100
            p.iloc[0] = 100
            prices[name] = p
    
    monthly = prices.resample('ME').last()
    monthly_dca = dca_total / int(dca_months)
    
    results = []
    
    for name in selected_portfolios:
        lump_units = lump_sum / monthly[name].iloc[0]
        lump_final = lump_units * monthly[name].iloc[-1]
        
        dca_units = 0
        for i in range(min(int(dca_months), len(monthly))):
            dca_units += monthly_dca / monthly[name].iloc[i]
        dca_final = dca_units * monthly[name].iloc[-1]
        
        lump_return = (lump_final / lump_sum - 1) * 100
        dca_return = (dca_final / dca_total - 1) * 100
        
        results.append({
            'Portfolio': name,
            'Lump Sum Return': f"{lump_return:.2f}%",
            'DCA Return': f"{dca_return:.2f}%",
            'Winner': '🏆 Lump Sum' if lump_return > dca_return else '🏆 DCA'
        })
    
    fig = go.Figure()
    portfolios_list = [r['Portfolio'] for r in results]
    lump_returns = [float(r['Lump Sum Return'].replace('%', '')) for r in results]
    dca_returns = [float(r['DCA Return'].replace('%', '')) for r in results]
    
    fig.add_trace(go.Bar(name='Lump Sum', x=portfolios_list, y=lump_returns, marker_color='#1E90FF'))
    fig.add_trace(go.Bar(name='DCA', x=portfolios_list, y=dca_returns, marker_color='#228B22'))
    
    fig.update_layout(
        title=f'Lump Sum (₪{lump_sum:,.0f}) vs DCA (₪{dca_total:,.0f} over {int(dca_months)} months)',
        xaxis_title='Portfolio', yaxis_title='Return (%)',
        barmode='group', template='plotly_white'
    )
    
    return fig, pd.DataFrame(results)


# ============ 9. MASTER SIMULATION ============
def master_simulation():
    """Run comprehensive analysis."""
    portfolios = ['S&P 500', 'Nasdaq 100', 'Semiconductors', 'Golden Butterfly']
    
    all_tickers = [USD_ILS_TICKER]
    for name in portfolios:
        all_tickers.extend(PORTFOLIOS[name])
    all_tickers = list(set(all_tickers))
    
    data = yf.download(all_tickers, start="2005-01-01", progress=False, auto_adjust=True)['Close']
    data = data.ffill().dropna()
    
    usd_ils = data[USD_ILS_TICKER]
    cpi = build_cpi_index(data.index)
    
    prices_nom = pd.DataFrame(index=data.index)
    prices_real = pd.DataFrame(index=data.index)
    
    for name in portfolios:
        tickers = PORTFOLIOS[name]
        if len(tickers) == 1:
            p = data[tickers[0]] * usd_ils
            norm_p = (p / p.iloc[0]) * 100
        else:
            basket = data[tickers].multiply(usd_ils, axis=0)
            rets = basket.pct_change().mean(axis=1)
            norm_p = (1 + rets).cumprod() * 100
            norm_p.iloc[0] = 100
        
        prices_nom[name] = norm_p
        prices_real[name] = norm_p / cpi * 100
    
    fig = make_subplots(rows=2, cols=2,
                        subplot_titles=['Nominal Returns (ILS)', 'Real Returns (Inflation-Adjusted)',
                                        'Nominal Returns (Log Scale)', 'Israeli CPI Index'])
    
    for name in portfolios:
        color = COLORS.get(name, '#888888')
        fig.add_trace(go.Scatter(x=data.index, y=prices_nom[name], name=name,
                                 line=dict(color=color, width=2)), row=1, col=1)
        fig.add_trace(go.Scatter(x=data.index, y=prices_real[name], name=name,
                                 line=dict(color=color, width=2), showlegend=False), row=1, col=2)
        fig.add_trace(go.Scatter(x=data.index, y=prices_nom[name], name=name,
                                 line=dict(color=color, width=2), showlegend=False), row=2, col=1)
    
    fig.add_trace(go.Scatter(x=data.index, y=cpi, name='CPI',
                             line=dict(color='red', width=2)), row=2, col=2)
    
    fig.update_yaxes(type='log', row=2, col=1)
    fig.add_hline(y=100, line_dash="dash", line_color="gray", opacity=0.5)
    
    fig.update_layout(
        title='Master Simulation: Complete Portfolio Analysis (2005 - Present)',
        hovermode='x unified', template='plotly_white', height=700, showlegend=True
    )
    
    results = []
    for name in portfolios:
        results.append({
            'Portfolio': name,
            'Nominal Return': f"{prices_nom[name].iloc[-1] - 100:.2f}%",
            'Real Return': f"{prices_real[name].iloc[-1] - 100:.2f}%"
        })
    
    results.append({
        'Portfolio': '📊 Total Inflation',
        'Nominal Return': f"{cpi.iloc[-1] - 100:.2f}%",
        'Real Return': '-'
    })
    
    return fig, pd.DataFrame(results)


print("✅ All analysis functions loaded!")
```
