```python
# @title
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def analyze_silver_returns():
    # 1. הגדרת התקופה והנכסים
    # SLV = תעודת סל כסף
    # ILS=X = שער הדולר מול השקל
    tickers = ['SLV', 'ILS=X']

    print("מוריד נתונים פיננסיים...")
    data = yf.download(tickers, period="1y", progress=False)['Close']

    # ניקוי נתונים (מחיקת ימים ללא מסחר ושיטת Forward Fill לחוסרים)
    data = data.ffill().dropna()

    # 2. שימוש ב-Numpy לביצוע מניפולציות ווקטוריות
    # המרת עמודות ה-Pandas למערכי Numpy לביצועים מתמטיים מהירים
    slv_usd_arr = data['SLV'].to_numpy()
    usd_ils_arr = data['ILS=X'].to_numpy()

    # חישוב ווקטורי של מחיר הכסף בשקלים (סימולציה של הקרן הישראלית)
    # הכפלה איבר-באיבר (Element-wise multiplication)
    slv_ils_synthetic_arr = np.multiply(slv_usd_arr, usd_ils_arr)

    # 3. חישוב תשואות לוגריתמיות (Log Returns) באמצעות Numpy
    # שימוש בלוגריתם טבעי הוא מדויק יותר מתמטית לחישוב תשואה מצטברת
    # הנוסחה: ln(Pt / Pt-1)

    # SLV בדולר
    log_returns_usd = np.diff(np.log(slv_usd_arr))
    # SLV בשקלים (הסימולציה)
    log_returns_ils = np.diff(np.log(slv_ils_synthetic_arr))

    # 4. חישוב תשואה מצטברת (Cumulative Return)
    # סכימה מצטברת של התשואות הלוגריתמיות והמרתן חזרה לאחוזים (exp)
    cumulative_usd = np.exp(np.cumsum(log_returns_usd)) - 1
    cumulative_ils = np.exp(np.cumsum(log_returns_ils)) - 1

    # הוספת אפס להתחלה כדי שהגרף יתחיל מנקודה 0
    cumulative_usd = np.insert(cumulative_usd, 0, 0)
    cumulative_ils = np.insert(cumulative_ils, 0, 0)

    # 5. סטטיסטיקה בסיסית עם Numpy
    volatility_usd = np.std(log_returns_usd) * np.sqrt(252) * 100 # שנתית
    volatility_ils = np.std(log_returns_ils) * np.sqrt(252) * 100 # שנתית
    correlation = np.corrcoef(log_returns_usd, log_returns_ils)[0, 1]

    # הדפסת נתונים
    print(f"\n--- תוצאות ניתוח שנתי ---")
    print(f"תשואה שנתית SLV (דולרי): {cumulative_usd[-1]*100:.2f}%")
    print(f"תשואה שנתית SLV (מותאם לשקל): {cumulative_ils[-1]*100:.2f}%")
    print(f"הפער שנוצר בגלל המטבע: {(cumulative_ils[-1] - cumulative_usd[-1])*100:.2f}%")
    print(f"סטיית תקן (סיכון) דולרית: {volatility_usd:.2f}%")
    print(f"סטיית תקן (סיכון) שקלית: {volatility_ils:.2f}%")
    print(f"קורלציה בין התשואות: {correlation:.4f}")

    # 6. ויזואליזציה
    dates = data.index
    plt.figure(figsize=(12, 6))

    # המרת התשואות לאחוזים לתצוגה
    plt.plot(dates, cumulative_usd * 100, label='iShares SLV (USD)', color='blue', linewidth=2)
    plt.plot(dates, cumulative_ils * 100, label='Simulated Shekel Silver (ILS)', color='green', linestyle='--', linewidth=2)

    plt.title('Silver Return Comparison: USD vs. ILS Adjusted (1 Year)')
    plt.xlabel('Date')
    plt.ylabel('Cumulative Return (%)')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.show()

if __name__ == "__main__":
    analyze_silver_returns()
```

```python
# @title
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def analyze_sp500_synthetic_il():
    # IVV = S&P 500 ETF (המקור האמריקאי)
    # ILS=X = שער הדולר/שקל
    tickers = ['IVV', 'ILS=X']

    print("מוריד נתונים ל-5 שנים (S&P 500 + שער דולר)...")
    # הורדת נתונים
    data = yf.download(tickers, period="5y", progress=False)['Close']

    # ניקוי נתונים: מחיקת שורות ריקות והשלמת נתונים חסרים (סופ"ש וחגים)
    data = data.ffill().dropna()

    if data.empty:
        print("שגיאה: לא התקבלו נתונים. נסה שוב מאוחר יותר.")
        return

    # המרה ל-Numpy לביצועים מהירים
    sp500_usd_arr = data['IVV'].to_numpy()
    usd_ils_arr = data['ILS=X'].to_numpy()

    # --- יצירת הקרן הישראלית הסינתטית ---
    # הכפלה של המדד האמריקאי בשער הדולר נותנת את השווי השקלי המדויק
    # זה מדמה בדיוק את מה שקורה בקרן "תכלית" או "קסם" לא מנוטרלת
    sp500_ils_synthetic = np.multiply(sp500_usd_arr, usd_ils_arr)

    # --- חישוב תשואות לוגריתמיות ---
    log_ret_usd = np.diff(np.log(sp500_usd_arr))
    log_ret_ils = np.diff(np.log(sp500_ils_synthetic))

    # --- חישוב תשואה מצטברת ---
    cum_ret_usd = np.exp(np.cumsum(log_ret_usd)) - 1
    cum_ret_ils = np.exp(np.cumsum(log_ret_ils)) - 1

    # איפוס גרפים לנקודת התחלה 0
    cum_ret_usd = np.insert(cum_ret_usd, 0, 0)
    cum_ret_ils = np.insert(cum_ret_ils, 0, 0)

    # --- סטטיסטיקה ---
    vol_usd = np.std(log_ret_usd) * np.sqrt(252) * 100
    vol_ils = np.std(log_ret_ils) * np.sqrt(252) * 100

    # פער סופי באחוזים
    gap = (cum_ret_ils[-1] - cum_ret_usd[-1]) * 100

    print(f"\n--- תוצאות: S&P 500 דולרי מול שקלי (5 שנים) ---")
    print(f"תשואה מצטברת דולרית (IVV): {cum_ret_usd[-1]*100:.2f}%")
    print(f"תשואה מצטברת שקלית (סימולציה לקרן ישראלית): {cum_ret_ils[-1]*100:.2f}%")
    print(f"הפער שנוצר עקב שינויי מטבע: {gap:+.2f}%")
    print(f"סטיית תקן (סיכון) דולרית: {vol_usd:.2f}%")
    print(f"סטיית תקן (סיכון) שקלית: {vol_ils:.2f}%")

    # --- ויזואליזציה ---
    dates = data.index
    plt.figure(figsize=(12, 6))

    plt.plot(dates, cum_ret_usd * 100, label='S&P 500 (USD Base)', color='blue', linewidth=1.5)
    plt.plot(dates, cum_ret_ils * 100, label='S&P 500 (Shekel Value)', color='green', linestyle='--', linewidth=1.5)

    plt.title('S&P 500 Return: USD vs. ILS Holder (5 Years)')
    plt.xlabel('Date')
    plt.ylabel('Cumulative Return (%)')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.show()

if __name__ == "__main__":
    analyze_sp500_synthetic_il()
```

```python
# @title
import requests
import pandas as pd
import yfinance as yf
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta

def get_tase_history(security_id, years=5):
    """
    פונקציה לשליפת נתונים היסטוריים ישירות מה-API של הבורסה בתל אביב
    """
    # כתובת ה-API הפנימית של הבורסה לנתוני גרף
    url = "https://api.tase.co.il/api/Chart/History"

    # חישוב תאריכים
    today = datetime.now()
    start_date = today - timedelta(days=years*365)

    # הפרמטרים שה-API מצפה לקבל
    params = {
        "pType": 1,         # סוג נייר (1 בדרך כלל לתעודות/מניות)
        "id": security_id,  # מספר הנייר (למשל 1159250)
        "l": 4,             # סוג גרף (4 = מותאם דיבידנד וריבית)
        "d": 0              # מביא את כל ההיסטוריה הזמינה
    }

    # Headers חובה - אחרת הבורסה תחסום את הבקשה
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36",
        "Referer": "https://www.tase.co.il/",
        "Accept-Language": "he-IL,he;q=0.9,en-US;q=0.8,en;q=0.7"
    }

    print(f"מתחבר לשרתי הבורסה לשליפת נייר {security_id}...")
    try:
        response = requests.get(url, params=params, headers=headers)
        response.raise_for_status()
        data = response.json()

        # המרת ה-JSON ל-DataFrame
        if 'c' in data: # 'c' מכיל את רשימת הנרות (Candles)
            df = pd.DataFrame(data['c'])
            # d = Date, c = Close price
            df = df[['d', 'c']]
            df.columns = ['Date', 'Close']

            # המרת תאריך מפורמט מוזר של הבורסה ל-Datetime
            # הבורסה מחזירה תאריך כסטרינג "/Date(1609459200000)/"
            df['Date'] = df['Date'].astype(str).str.extract('(\d+)').astype(int)
            df['Date'] = pd.to_datetime(df['Date'], unit='ms')

            df.set_index('Date', inplace=True)

            # סינון לפי השנים שביקשנו
            df = df[df.index >= start_date]

            # המרה מאגורות לשקלים (נתוני הבורסה הם באגורות)
            df['Close'] = df['Close'] / 100

            return df
        else:
            print("המבנה של ה-JSON השתנה או שאין נתונים.")
            return None

    except Exception as e:
        print(f"שגיאה בשליפת נתונים מהבורסה: {e}")
        return None

def compare_real_tase_vs_sp500():
    # 1. שליפת נתונים מהבורסה הישראלית (תכלית S&P 500)
    # מספר נייר: 1159250
    il_df = get_tase_history(security_id="1159250", years=5)

    if il_df is None or il_df.empty:
        return

    # 2. שליפת נתונים מיאהו (S&P 500 אמריקאי)
    print("מוריד נתונים מארה\"ב (Yahoo)...")
    us_tickers = ['IVV']
    us_df = yf.download(us_tickers, start=il_df.index[0], progress=False)['Close']

    # 3. איחוד הנתונים (Merge)
    # שימוש ב-merge_asof או reindex כדי לסנכרן תאריכים
    combined = pd.DataFrame(index=il_df.index)
    combined['TASE_Fund'] = il_df['Close']

    # התאמת הנתונים האמריקאים לתאריכים הישראלים (מילוי חוסרים)
    us_df_reindexed = us_df['IVV'].reindex(il_df.index, method='ffill')
    combined['US_ETF_USD'] = us_df_reindexed

    # ניקוי שורות ריקות שנוצרו בהתחלה
    combined = combined.dropna()

    # 4. חישוב תשואות (Numpy)
    tase_arr = combined['TASE_Fund'].to_numpy()
    us_arr = combined['US_ETF_USD'].to_numpy()

    # תשואה מצטברת
    # אנחנו משווים תשואה באחוזים, אז המטבע (שקל/דולר) לא משנה לחישוב השינוי היחסי
    cum_ret_tase = (tase_arr / tase_arr[0]) - 1
    cum_ret_us = (us_arr / us_arr[0]) - 1

    # ויזואליזציה
    plt.figure(figsize=(12, 6))
    plt.plot(combined.index, cum_ret_us * 100, label='S&P 500 (IVV) - USD Base', color='blue')
    plt.plot(combined.index, cum_ret_tase * 100, label='Tachlit Israel (1159250) - ILS Base', color='red')

    plt.title('Real Data: TASE API vs Yahoo Finance (5 Years)')
    plt.ylabel('Cumulative Return (%)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

    print(f"תשואה סופית קרן ישראלית: {cum_ret_tase[-1]*100:.2f}%")
    print(f"תשואה סופית קרן אמריקאית (דולרי): {cum_ret_us[-1]*100:.2f}%")

if __name__ == "__main__":
    compare_real_tase_vs_sp500()
```

```python
# @title
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
import io
from google.colab import files

def analyze_uploaded_file():
    print("שלב 1: אנא בחר את קובץ ה-CSV מהמחשב שלך...")
    # פקודה זו פותחת את כפתור ההעלאה של גוגל
    uploaded = files.upload()

    # שליפת שם הקובץ שהעלית (לא משנה איך קוראים לו)
    if not uploaded:
        print("לא הועלה קובץ. אנא נסה שנית.")
        return

    file_name = next(iter(uploaded))
    print(f"הקובץ '{file_name}' נטען בהצלחה! מעבד נתונים...")

    # קריאת הקובץ מתוך הזיכרון
    try:
        df = pd.read_csv(io.BytesIO(uploaded[file_name]))
    except Exception as e:
        print(f"שגיאה בקריאת הקובץ: {e}")
        return

    # --- מכאן זה אותו ניתוח כמו קודם ---

    # סידור עמודות ותאריכים
    # נניח שהקובץ בעברית כמו שראינו: 'תאריך' ו-'מחיר פדיון'
    # אם השמות שונים, הקוד ינסה לזהות לבד
    if 'תאריך' in df.columns:
        date_col = 'תאריך'
        price_col = 'מחיר פדיון'
    elif 'Date' in df.columns:
        date_col = 'Date'
        price_col = 'Price' # או Close, נתאים בהמשך
    else:
        # ניסיון ניחוש לפי מיקום (עמודה 1 תאריך, עמודה 3 מחיר)
        date_col = df.columns[1]
        price_col = df.columns[3]

    # המרת שמות
    df = df.rename(columns={date_col: 'Date', price_col: 'IL_Price'})

    # המרת תאריך (DD.MM.YYYY)
    df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
    df = df.dropna(subset=['Date']) # מחיקת שורות ללא תאריך
    df = df.sort_values('Date').set_index('Date')

    # חישוב תשואה מצטברת לקרן (מתחיל מ-0)
    # המרה למספרים (למקרה שיש פסיקים)
    if df['IL_Price'].dtype == object:
         df['IL_Price'] = df['IL_Price'].astype(str).str.replace(',', '').astype(float)

    df['IL_Return'] = (df['IL_Price'] / df['IL_Price'].iloc[0] - 1) * 100

    # הורדת S&P 500 להשוואה
    start_date = df.index[0]
    end_date = df.index[-1]

    print(f"מוריד נתוני S&P 500 לתקופה: {start_date.date()} עד {end_date.date()}...")
    us_data = yf.download('IVV', start=start_date, end=end_date + pd.Timedelta(days=5), progress=False)['Close']

    # איחוד נתונים
    combined = pd.merge_asof(df[['IL_Return']], us_data, left_index=True, right_index=True, direction='backward')
    combined = combined.rename(columns={'IVV': 'US_Price'}) # לפעמים יורד כטיקר אחר, נסדר ידנית אם צריך

    # טיפול במקרה ש-yfinance מחזיר מבנה שונה
    if 'US_Price' not in combined.columns and not us_data.empty:
        combined['US_Price'] = us_data

    combined = combined.dropna()
    combined['US_Return'] = (combined['US_Price'] / combined['US_Price'].iloc[0] - 1) * 100

    # גרף
    plt.figure(figsize=(12, 6))
    plt.plot(combined.index, combined['US_Return'], label='S&P 500 (USD)', color='blue')
    plt.plot(combined.index, combined['IL_Return'], label='Your Fund (ILS)', color='red')

    gap = combined['IL_Return'].iloc[-1] - combined['US_Return'].iloc[-1]

    plt.title(f'Fund vs S&P 500 Performance\nGap: {gap:.2f}%')
    plt.ylabel('Return (%)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

    print(f"תשואה סופית קרן: {combined['IL_Return'].iloc[-1]:.2f}%")
    print(f"תשואה סופית מדד: {combined['US_Return'].iloc[-1]:.2f}%")

if __name__ == "__main__":
    analyze_uploaded_file()
```

```python
# @title
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
import io
from google.colab import files
from datetime import timedelta

def analyze_last_year():
    print("שלב 1: אנא בחר את קובץ ה-CSV מהמחשב שלך...")
    uploaded = files.upload()

    if not uploaded:
        print("לא הועלה קובץ.")
        return

    file_name = next(iter(uploaded))

    # קריאת הקובץ
    try:
        df = pd.read_csv(io.BytesIO(uploaded[file_name]))
    except:
        print("שגיאה בקריאת הקובץ")
        return

    # --- סידור הנתונים ---
    # זיהוי עמודות
    if 'תאריך' in df.columns:
        date_col = 'תאריך'
        price_col = 'מחיר פדיון'
    elif 'Date' in df.columns:
        date_col = 'Date'
        price_col = 'Price'
    else:
        date_col = df.columns[1]
        price_col = df.columns[3]

    df = df.rename(columns={date_col: 'Date', price_col: 'IL_Price'})
    df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
    df = df.dropna(subset=['Date']).sort_values('Date').set_index('Date')

    # ניקוי פסיקים במחיר אם יש
    if df['IL_Price'].dtype == object:
         df['IL_Price'] = df['IL_Price'].astype(str).str.replace(',', '').astype(float)

    # --- החלק החדש: סינון לשנה האחרונה בלבד ---
    last_available_date = df.index[-1] # התאריך הכי עדכני בקובץ
    start_date_1y = last_available_date - timedelta(days=365) # שנה אחורה

    # חיתוך הדאטה-פריים (ניקח רק שורות שהן החל מלפני שנה)
    df = df[df.index >= start_date_1y]

    print(f"טווח התאריכים לניתוח: {df.index[0].date()} עד {df.index[-1].date()}")

    # חישוב תשואה (מתאפס ל-0 ביום הראשון של השנה הזו)
    df['IL_Return'] = (df['IL_Price'] / df['IL_Price'].iloc[0] - 1) * 100

    # --- השוואה ל-S&P 500 באותה תקופה ---
    print("מוריד נתוני S&P 500 לתקופה המקבילה...")
    us_data = yf.download('IVV', start=df.index[0], end=df.index[-1] + timedelta(days=5), progress=False)['Close']

    combined = pd.merge_asof(df[['IL_Return']], us_data, left_index=True, right_index=True, direction='backward')

    # טיפול במבנה הנתונים של yfinance
    if isinstance(combined, pd.DataFrame) and 'IVV' in combined.columns:
         combined = combined.rename(columns={'IVV': 'US_Price'})
    elif not us_data.empty:
         combined['US_Price'] = us_data

    combined = combined.dropna()
    combined['US_Return'] = (combined['US_Price'] / combined['US_Price'].iloc[0] - 1) * 100

    # גרף
    plt.figure(figsize=(12, 6))
    plt.plot(combined.index, combined['US_Return'], label='S&P 500 (USD)', color='blue')
    plt.plot(combined.index, combined['IL_Return'], label='Your Fund (ILS)', color='red')

    gap = combined['IL_Return'].iloc[-1] - combined['US_Return'].iloc[-1]

    plt.title(f'Performance: Last 365 Days\nGap: {gap:.2f}%')
    plt.ylabel('Return (%)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

    print(f"תשואה שנה אחרונה (קרן): {combined['IL_Return'].iloc[-1]:.2f}%")
    print(f"תשואה שנה אחרונה (מדד): {combined['US_Return'].iloc[-1]:.2f}%")

if __name__ == "__main__":
    analyze_last_year()
```

```python
# @title
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

def compare_portfolios_shekel_20y():
    # 1. הגדרת הנכסים (ETFs אמריקאיות שמייצגות את האפיקים)
    # ---------------------------------------------------------
    # Golden Butterfly:
    # VTI = מניות כללי | VBR = מניות ערך קטנות | TLT = אג"ח ארוך | SHY = אג"ח קצר | GLD = זהב
    golden_tickers = ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']

    # Ivy Portfolio:
    # VTI = מניות ארה"ב | VEU = מניות עולם | AGG = אג"ח כללי | DBC = סחורות | VNQ = נדל"ן
    # הערה: נשתמש ב-AGG במקום BND כי יש לו יותר היסטוריה, וב-VEU למניות עולם
    ivy_tickers = ['VTI', 'VEU', 'AGG', 'DBC', 'VNQ']

    # איחוד כל הטיקרים לרשימה אחת למשיכה יעילה + שער הדולר
    all_tickers = list(set(golden_tickers + ivy_tickers + ['ILS=X']))

    print("מוריד נתונים היסטוריים מ-20 השנים האחרונות (החל מ-2006)...")
    # DBC (סחורות) הושק ב-2006, ולכן ההיסטוריה המשותפת תתחיל משם (כ-19-20 שנה)
    data = yf.download(all_tickers, start="2006-01-01", progress=False)['Close']

    # מילוי נתונים חסרים (אם יום אחד לא נסחר)
    data = data.ffill().dropna()

    # 2. המרה לשקלים (הצמדה למטבע)
    # ---------------------------------------------------------
    # לוקחים כל נכס ומכפילים אותו בשער הדולר של אותו יום
    usd_ils = data['ILS=X']
    data_ils = data.copy()

    for ticker in all_tickers:
        if ticker != 'ILS=X':
            data_ils[ticker] = data[ticker] * usd_ils

    # 3. חישוב תשואות יומיות
    # ---------------------------------------------------------
    # אנו מחשבים את אחוז השינוי היומי של כל נכס בשקלים
    returns = data_ils.pct_change().dropna()

    # 4. בניית התיקים (איזון יומי - 20% לכל נכס)
    # ---------------------------------------------------------
    # הנחה: התיק "מאוזן" (Rebalanced) - זה נותן את הייצוג הטוב ביותר לאסטרטגיה

    # חישוב התשואה היומית של כל תיק (ממוצע פשוט של 5 הנכסים)
    golden_daily = returns[golden_tickers].mean(axis=1)
    ivy_daily = returns[ivy_tickers].mean(axis=1)

    # חישוב תשואה מצטברת (Compounded Returns)
    golden_cum = (1 + golden_daily).cumprod() - 1
    ivy_cum = (1 + ivy_daily).cumprod() - 1

    # 5. ויזואליזציה
    # ---------------------------------------------------------
    plt.figure(figsize=(12, 6))

    # המרה לאחוזים
    plt.plot(golden_cum.index, golden_cum * 100, label='Golden Butterfly (Shekel)', color='#D4AF37', linewidth=2) # צבע זהב
    plt.plot(ivy_cum.index, ivy_cum * 100, label='Ivy Portfolio (Shekel)', color='green', linewidth=2)

    # חישוב נתונים סופיים להצגה
    total_golden = golden_cum.iloc[-1] * 100
    total_ivy = ivy_cum.iloc[-1] * 100

    plt.title(f'20-Year Comparison: Golden Butterfly vs Ivy (ILS Adjusted)\nGolden: {total_golden:.0f}% | Ivy: {total_ivy:.0f}%', fontsize=14)
    plt.xlabel('Year')
    plt.ylabel('Cumulative Return (%)')
    plt.legend()
    plt.grid(True, alpha=0.3)

    # סימון משברים (2008, 2020, 2022)
    plt.axvspan(pd.Timestamp('2008-01-01'), pd.Timestamp('2009-01-01'), color='gray', alpha=0.1, label='2008 Crisis')

    print(f"--- סיכום ביצועים בשקלים (2006-היום) ---")
    print(f"תיק פרפר הזהב: {total_golden:.2f}%")
    print(f"תיק הקיסוס:   {total_ivy:.2f}%")

    plt.show()

if __name__ == "__main__":
    compare_portfolios_shekel_20y()
```

```python
# @title
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def compare_portfolios_with_inflation():
    # 1. הגדרת התיקים
    # ---------------------------------------------------------
    # Golden Butterfly: 20% לכל אחד
    golden_tickers = ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']

    # Ivy Portfolio: 20% לכל אחד (הערה: VEU הושק ב-2007, אז נתחיל משם)
    ivy_tickers = ['VTI', 'VEU', 'AGG', 'DBC', 'VNQ']

    all_tickers = list(set(golden_tickers + ivy_tickers + ['ILS=X']))

    print("1. מוריד נתונים היסטוריים (מניות, אג\"ח, זהב, סחורות ומט\"ח)...")
    # מתחילים מ-2007 כי זה התאריך המוקדם ביותר המשותף לכל הניירות (בעיקר VEU)
    data = yf.download(all_tickers, start="2007-01-01", progress=False)['Close']
    data = data.ffill().dropna()

    # 2. המרה לשקלים (נטרול השפעת המטבע)
    # ---------------------------------------------------------
    usd_ils = data['ILS=X']
    data_ils = pd.DataFrame(index=data.index)

    for ticker in all_tickers:
        if ticker != 'ILS=X':
            # שווי הנכס בשקלים = מחירו בדולר * שער הדולר
            data_ils[ticker] = data[ticker] * usd_ils

    # חישוב תשואות יומיות בשקלים
    returns_ils = data_ils.pct_change().dropna()

    # 3. בניית התיקים (איזון יומי - 20% לכל נכס)
    # ---------------------------------------------------------
    golden_daily = returns_ils[golden_tickers].mean(axis=1)
    ivy_daily = returns_ils[ivy_tickers].mean(axis=1)

    # תשואה מצטברת (מתחילים מ-100 נקודות)
    golden_cum = (1 + golden_daily).cumprod() * 100
    ivy_cum = (1 + ivy_daily).cumprod() * 100

    # 4. הוספת נתוני אינפלציה (ישראל)
    # ---------------------------------------------------------
    # נתונים היסטוריים (שנתיים) של המדד בישראל (מקור: למ"ס/גלובס/ביזפורטל)
    inflation_rates = {
        2007: 3.4, 2008: 3.8, 2009: 3.9, 2010: 2.7,
        2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8,
        2019: 0.6, 2020: -0.7, 2021: 2.8, 2022: 5.3,
        2023: 3.0, 2024: 3.0 # הערכה ל-2024
    }

    # יצירת קו אינפלציה יומי (אינטרפולציה לינארית כדי שיראה יפה בגרף)
    cpi_series = pd.Series(index=golden_cum.index, dtype=float)
    cpi_series[:] = np.nan

    # נקודת התחלה
    current_cpi = 100
    cpi_series.iloc[0] = current_cpi

    # חישוב הערך בסוף כל שנה
    years = sorted(inflation_rates.keys())
    for year in years:
        rate = inflation_rates[year] / 100
        current_cpi = current_cpi * (1 + rate)
        # מציאת התאריך האחרון באותה שנה בדאטה
        end_of_year = data.index[data.index.year == year][-1]
        cpi_series.loc[end_of_year] = current_cpi

    # השלמת הפערים (קו ישר בין השנים)
    cpi_series = cpi_series.interpolate(method='time')

    # נרמול התיקים שיתחילו באותה נקודה כמו האינפלציה (100) ביום הראשון המשותף
    start_val = cpi_series.dropna().iloc[0] # הערך ב-2007

    # חיתוך הנתונים לתקופה שיש לה אינפלציה
    valid_dates = cpi_series.dropna().index
    start_date = valid_dates[0]

    golden_plot = golden_cum[start_date:]
    ivy_plot = ivy_cum[start_date:]
    cpi_plot = cpi_series[start_date:]

    # איפוס כולם ל-100 בנקודת ההתחלה
    golden_plot = golden_plot / golden_plot.iloc[0] * 100
    ivy_plot = ivy_plot / ivy_plot.iloc[0] * 100
    cpi_plot = cpi_plot / cpi_plot.iloc[0] * 100

    # 5. ויזואליזציה
    # ---------------------------------------------------------
    plt.figure(figsize=(12, 7))

    plt.plot(golden_plot, label='Golden Butterfly (ILS)', color='#D4AF37', linewidth=2.5) # זהב
    plt.plot(ivy_plot, label='Ivy Portfolio (ILS)', color='green', linewidth=2)
    plt.plot(cpi_plot, label='Israel Inflation (CPI)', color='red', linestyle='--', linewidth=2, alpha=0.7)

    # חישוב תשואה סופית
    ret_golden = golden_plot.iloc[-1] - 100
    ret_ivy = ivy_plot.iloc[-1] - 100
    total_inflation = cpi_plot.iloc[-1] - 100

    plt.title(f'Portfolios vs. Israel Inflation (Real Terms in Shekels)\nInflation: +{total_inflation:.0f}% | Golden: +{ret_golden:.0f}% | Ivy: +{ret_ivy:.0f}%', fontsize=14)
    plt.xlabel('Year')
    plt.ylabel('Growth of 100 ILS')
    plt.legend()
    plt.grid(True, alpha=0.3)

    # מילוי השטח שבין האינפלציה לתיק (הרווח הריאלי)
    plt.fill_between(golden_plot.index, golden_plot, cpi_plot, where=(golden_plot > cpi_plot), color='#D4AF37', alpha=0.1, label='Real Profit')

    print(f"--- תוצאות בשקלים (2007-2024) ---")
    print(f"אינפלציה מצטברת בישראל: {total_inflation:.2f}% (יוקר המחיה עלה)")
    print(f"תיק פרפר הזהב: {ret_golden:.2f}% (פי {ret_golden/total_inflation:.1f} מהאינפלציה)")
    print(f"תיק הקיסוס:   {ret_ivy:.2f}%")

    plt.show()

if __name__ == "__main__":
    compare_portfolios_with_inflation()
```

```python
# @title
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def plot_real_shekel_yield():
    # 1. הגדרת התיקים (20% לכל נכס)
    # Golden Butterfly: מניות, מניות ערך, אג"ח ארוך, אג"ח קצר, זהב
    golden_tickers = ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']

    # Ivy Portfolio: מניות ארה"ב, מניות עולם, אג"ח, סחורות, נדל"ן
    # הערה: נתונים ל-VEU ו-DBC זמינים מ-2007 (כ-18 שנה)
    ivy_tickers = ['VTI', 'VEU', 'AGG', 'DBC', 'VNQ']

    all_tickers = list(set(golden_tickers + ivy_tickers + ['ILS=X']))

    print("מוריד נתונים היסטוריים (מניות, אג\"ח, מט\"ח)...")
    try:
        data = yf.download(all_tickers, start="2007-01-01", progress=False)['Close']
        data = data.ffill().dropna()
    except Exception as e:
        print("שגיאה בהורדת נתונים. וודא שיש חיבור לאינטרנט.")
        return

    # 2. המרה לשקלים (Nominal ILS Return)
    usd_ils = data['ILS=X']
    data_ils = pd.DataFrame(index=data.index)

    for ticker in all_tickers:
        if ticker != 'ILS=X':
            data_ils[ticker] = data[ticker] * usd_ils

    # חישוב תשואות יומיות
    returns_ils = data_ils.pct_change().dropna()

    # 3. בניית התיקים (Nominal)
    golden_daily = returns_ils[golden_tickers].mean(axis=1)
    ivy_daily = returns_ils[ivy_tickers].mean(axis=1)

    # מדד נומינלי (ללא התחשבות באינפלציה) - מתחיל מ-100
    golden_nominal = (1 + golden_daily).cumprod() * 100
    ivy_nominal = (1 + ivy_daily).cumprod() * 100

    # 4. בניית מדד האינפלציה של ישראל (נתונים היסטוריים רשמיים)
    # המקור: הלשכה המרכזית לסטטיסטיקה
    inflation_data = {
        2007: 3.4, 2008: 3.8, 2009: 3.9, 2010: 2.7,
        2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8,
        2019: 0.6, 2020: -0.7, 2021: 2.8, 2022: 5.3,
        2023: 3.0, 2024: 3.0  # הערכה
    }

    # יצירת "מדד אינפלציה מצטבר" יומי
    cpi_series = pd.Series(index=golden_nominal.index, dtype=float)
    cpi_series[:] = np.nan
    current_cpi = 100
    cpi_series.iloc[0] = 100

    for year in sorted(inflation_data.keys()):
        if year in data.index.year:
            rate = inflation_data[year] / 100
            current_cpi = current_cpi * (1 + rate)
            # הצמדה לסוף השנה
            last_day = data.index[data.index.year == year][-1]
            if last_day in cpi_series.index:
                cpi_series.loc[last_day] = current_cpi

    cpi_series = cpi_series.interpolate(method='time') # השלמת ימים חסרים בקו ישר

    # 5. חישוב התשואה הריאלית (Real Return)
    # הנוסחה: (מדד התיק / מדד האינפלציה) * 100

    # חיתוך לתקופה המשותפת
    start_date = cpi_series.first_valid_index()
    golden_plot = golden_nominal.loc[start_date:]
    ivy_plot = ivy_nominal.loc[start_date:]
    cpi_plot = cpi_series.loc[start_date:]

    # חישוב ריאלי: מנכה את עליית המחירים
    golden_real = (golden_plot / cpi_plot) * 100
    ivy_real = (ivy_plot / cpi_plot) * 100

    # נרמול להתחלה מ-100
    golden_real = golden_real / golden_real.iloc[0] * 100
    ivy_real = ivy_real / ivy_real.iloc[0] * 100

    # 6. גרף
    plt.figure(figsize=(12, 7))

    plt.plot(golden_real, label='Golden Butterfly (Real Purchasing Power)', color='#D4AF37', linewidth=2.5)
    plt.plot(ivy_real, label='Ivy Portfolio (Real Purchasing Power)', color='green', linewidth=2)

    # קו האפס (ללא רווח ריאלי)
    plt.axhline(y=100, color='gray', linestyle='--', alpha=0.5, label='Zero Real Growth')

    total_golden = golden_real.iloc[-1] - 100
    total_ivy = ivy_real.iloc[-1] - 100

    plt.title(f'Real Shekel Return (Inflation Adjusted) 2007-2025\n(Net Profit in Purchasing Power)\nGolden: +{total_golden:.0f}% | Ivy: +{total_ivy:.0f}%')
    plt.xlabel('Year')
    plt.ylabel('Real Growth (Base=100)')
    plt.legend()
    plt.grid(True, alpha=0.3)

    print(f"--- תוצאות אמת ריאליות (בניכוי אינפלציה) ---")
    print(f"תיק פרפר הזהב: הגדיל את כוח הקנייה ב-{total_golden:.2f}%")
    print(f"תיק הקיסוס: הגדיל את כוח הקנייה ב-{total_ivy:.2f}%")
    print("המשמעות: זה הרווח ה'נקי' שלך אחרי ששילמת על ההתייקרויות בסופר ובדיור.")

    plt.show()

if __name__ == "__main__":
    plot_real_shekel_yield()
```

```python
# @title
import requests
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
import numpy as np

def plot_real_return_with_cbs_api():
    # --- שלב 1: משיכת נתוני אינפלציה (אמת) מהלמ"ס ---
    print("1. מתחבר ל-API של הלשכה המרכזית לסטטיסטיקה (מדד 120010)...")

    # כתובת ה-API לפי התיעוד ששלחת
    # מושכים נתונים מ-2005 כדי לכסות 20 שנה
    cbs_url = "https://api.cbs.gov.il/index/data/price"
    params = {
        "id": "120010",      # מדד המחירים לצרכן - כללי
        "format": "json",    # פורמט נוח לעבודה
        "startPeriod": "01-2005",
        "endPeriod": "12-2030",
        "download": "false"
    }

    try:
        response = requests.get(cbs_url, params=params)
        data_cbs = response.json()

        # הלמ"ס מחזיר רשימה של חודשים. אנו צריכים את ה-"percent" (השינוי החודשי)
        # כדי לבנות מדד רציף שלא נשבר כשמחליפים "שנת בסיס"
        if 'month' in data_cbs:
            records = data_cbs['month']
        else:
            # מבנה חליפי לפעמים
            records = data_cbs

        inflation_dates = []
        inflation_values = []

        # בניית מדד הנדסי (Cumulative Product)
        # מתחילים מ-100
        current_index = 100.0

        # הנתונים מגיעים לעיתים מהחדש לישן, נהפוך אותם כדי לחשב קדימה
        # (תלוי ב-API, ליתר ביטחון נמיין לפי תאריך)
        parsed_records = []
        for item in records:
            # חילוץ תאריך
            if 'date' in item and 'DateMonth' in item['date']:
                 dm = item['date']['DateMonth']
                 year = int(dm['year'])
                 month = int(dm['month'])
                 # השינוי החודשי באחוזים (למשל 0.2 או -0.1)
                 pct_change = float(dm['percent'])

                 date_obj = pd.Timestamp(f"{year}-{month}-01")
                 parsed_records.append({'date': date_obj, 'pct': pct_change})

        # מיון מהישן לחדש
        parsed_records.sort(key=lambda x: x['date'])

        # חישוב המדד המצטבר
        cpi_dates = [parsed_records[0]['date']]
        cpi_vals = [100.0]

        for rec in parsed_records:
            # הנוסחה: המדד החדש = המדד הקודם * (1 + השינוי/100)
            current_index = current_index * (1 + rec['pct'] / 100)
            # משייכים את המדד לסוף החודש (כי הוא מתפרסם ב-15 לחודש הבא)
            end_of_month = rec['date'] + pd.offsets.MonthEnd(0)
            cpi_dates.append(end_of_month)
            cpi_vals.append(current_index)

        # יצירת סדרת זמן יומית (אינטרפולציה)
        cpi_series = pd.Series(cpi_vals, index=cpi_dates)
        # הופכים מנתון חודשי ליומי (קו חלק)
        cpi_daily = cpi_series.resample('D').interpolate(method='time')

        print("V נתוני אינפלציה נמשכו בהצלחה וחושבו מחדש לרצף אחד.")

    except Exception as e:
        print(f"שגיאה במשיכת נתונים מהלמ\"ס: {e}")
        print("הערה: ייתכן וחומת האש חוסמת את הגישה ל-API של הממשלה.")
        return

    # --- שלב 2: משיכת נתוני בורסה (תיקי השקעות) ---
    print("2. מוריד נתוני שוק (מניות, זהב, אג\"ח) מארה\"ב...")

    # Golden Butterfly
    golden_tickers = ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']
    # Ivy Portfolio (VEU ו-DBC מתחילים באזור 2006-2007)
    ivy_tickers = ['VTI', 'VEU', 'AGG', 'DBC', 'VNQ']

    all_tickers = list(set(golden_tickers + ivy_tickers + ['ILS=X']))

    # משיכה מ-Yahoo
    market_data = yf.download(all_tickers, start="2007-01-01", progress=False)['Close']
    market_data = market_data.ffill().dropna()

    # --- שלב 3: חישוב התיקים (נומינלי בשקלים) ---
    usd_ils = market_data['ILS=X']
    data_ils = pd.DataFrame(index=market_data.index)

    for ticker in all_tickers:
        if ticker != 'ILS=X':
            data_ils[ticker] = market_data[ticker] * usd_ils

    # תשואות יומיות
    rets = data_ils.pct_change().dropna()

    # בניית המדדים (20% לכל נכס)
    golden_nom = (1 + rets[golden_tickers].mean(axis=1)).cumprod() * 100
    ivy_nom = (1 + rets[ivy_tickers].mean(axis=1)).cumprod() * 100

    # --- שלב 4: חישוב תשואה ריאלית (בניכוי המדד של הלמ"ס) ---
    print("3. מבצע הצמדה לאינפלציה (חישוב ריאלי)...")

    # חיתוך לתקופה המשותפת (איפה שיש לנו גם בורסה וגם למ"ס)
    common_start = max(golden_nom.index[0], cpi_daily.index[0])
    common_end = min(golden_nom.index[-1], cpi_daily.index[-1])

    # יישור הנתונים לאותו טווח תאריכים
    portfolio_g = golden_nom.loc[common_start:common_end]
    portfolio_i = ivy_nom.loc[common_start:common_end]
    cpi_adjusted = cpi_daily.reindex(portfolio_g.index).interpolate(method='time') # התאמה לימי מסחר

    # החישוב הריאלי: (שווי התיק / מדד המחירים)
    real_golden = (portfolio_g / cpi_adjusted)
    real_ivy = (portfolio_i / cpi_adjusted)

    # נרמול ל-100 ביום ההתחלה (כדי שיהיה קל להבין אחוזים)
    real_golden = real_golden / real_golden.iloc[0] * 100
    real_ivy = real_ivy / real_ivy.iloc[0] * 100

    # --- ויזואליזציה ---
    plt.figure(figsize=(12, 7))

    plt.plot(real_golden, label='Golden Butterfly (Real Purchasing Power)', color='#D4AF37', linewidth=2.5)
    plt.plot(real_ivy, label='Ivy Portfolio (Real Purchasing Power)', color='green', linewidth=2)

    # קו ה-0 (האם שמרנו על הערך?)
    plt.axhline(y=100, color='red', linestyle='--', alpha=0.5, label='Break-Even (Inflation Adjusted)')

    # חישוב תשואה סופית
    total_g = real_golden.iloc[-1] - 100
    total_i = real_ivy.iloc[-1] - 100

    plt.title(f'Real Returns (Net of Inflation) 2007-Today\nBased on CBS Index 120010 & Daily USD/ILS\nGolden: +{total_g:.1f}% | Ivy: +{total_i:.1f}% (Real Growth)', fontsize=14)
    plt.ylabel('Real Value (Base=100)')
    plt.xlabel('Year')
    plt.legend(loc='upper left')
    plt.grid(True, alpha=0.3)

    # מילוי שטח חיובי
    plt.fill_between(real_golden.index, real_golden, 100, where=(real_golden>100), color='#D4AF37', alpha=0.1)

    print(f"\n--- תוצאות אמת (כוח קנייה נטו) ---")
    print(f"תיק פרפר הזהב: הגדיל את כוח הקנייה שלך פי {real_golden.iloc[-1]/100:.2f}")
    print(f"תיק הקיסוס: הגדיל את כוח הקנייה שלך פי {real_ivy.iloc[-1]/100:.2f}")
    print("(משמעות: גם אחרי שהורדנו את התייקרות המחירים בישראל, עדיין הרווחת יפה).")

    plt.show()

if __name__ == "__main__":
    plot_real_return_with_cbs_api()
```

```python
# @title
# התקנת ספריית המידע (רץ רק ב-Colab)
!pip install yfinance --quiet

import yfinance as yf
import pandas as pd
import numpy as np

def print_yearly_breakdown():
    print("--- טוען נתונים ומחשב תשואות... ---")

    # 1. נתוני אינפלציה היסטוריים (למ"ס)
    inflation_data = {
        2007: 3.4, 2008: 3.8, 2009: 3.9, 2010: 2.7,
        2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8,
        2019: 0.6, 2020: -0.7, 2021: 2.8, 2022: 5.3,
        2023: 3.0, 2024: 3.2, 2025: 2.4
    }

    # 2. משיכת נתונים
    golden_tickers = ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']
    ivy_tickers = ['VTI', 'VEU', 'AGG', 'DBC', 'VNQ']
    all_tickers = list(set(golden_tickers + ivy_tickers + ['ILS=X']))

    try:
        # הורדת נתונים מ-2007
        data = yf.download(all_tickers, start="2007-01-01", progress=False)['Close']
        data = data.ffill().dropna()
    except Exception as e:
        print(f"שגיאה בהורדת נתונים: {e}")
        return

    # 3. בניית מדד האינפלציה היומי
    cpi_daily = pd.Series(index=data.index, dtype=float)
    cpi_daily[:] = np.nan
    cpi_daily.iloc[0] = 100.0
    current_cpi = 100.0

    for year in sorted(inflation_data.keys()):
        if year >= data.index[0].year:
            rate = inflation_data[year] / 100
            current_cpi = current_cpi * (1 + rate)
            # מציאת יום אחרון בשנה
            days_in_year = data.index[data.index.year == year]
            if not days_in_year.empty:
                cpi_daily.loc[days_in_year[-1]] = current_cpi

    cpi_daily = cpi_daily.interpolate(method='time').ffill()

    # 4. המרה לשקלים וחישוב תיקים
    usd_ils = data['ILS=X']
    data_ils = pd.DataFrame(index=data.index)
    for ticker in all_tickers:
        if ticker != 'ILS=X':
            data_ils[ticker] = data[ticker] * usd_ils

    rets = data_ils.pct_change().dropna()

    golden_nom = (1 + rets[golden_tickers].mean(axis=1)).cumprod() * 100
    ivy_nom = (1 + rets[ivy_tickers].mean(axis=1)).cumprod() * 100

    # 5. חישוב ריאלי
    start_date = max(golden_nom.index[0], cpi_daily.first_valid_index())

    p_golden = golden_nom.loc[start_date:]
    p_ivy = ivy_nom.loc[start_date:]
    p_cpi = cpi_daily.loc[start_date:]

    real_golden = (p_golden / p_cpi)
    real_ivy = (p_ivy / p_cpi)

    # איפוס ל-100
    real_golden = real_golden / real_golden.iloc[0] * 100
    real_ivy = real_ivy / real_ivy.iloc[0] * 100

    # ==========================================
    # הדפסת הטבלה השנתית
    # ==========================================
    print("\n" + "="*60)
    print(f"{'Year':<6} | {'Golden Butterfly (Real)':<24} | {'Ivy Portfolio (Real)':<24}")
    print("="*60)

    years = sorted(list(set(real_golden.index.year)))
    prev_g = 100.0
    prev_i = 100.0

    for year in years:
        # בדיקה שיש נתונים לאותה שנה
        subset_g = real_golden[real_golden.index.year == year]
        subset_i = real_ivy[real_ivy.index.year == year]

        if subset_g.empty: continue

        # הערך בסוף השנה
        curr_g = subset_g.iloc[-1]
        curr_i = subset_i.iloc[-1]

        # חישוב התשואה השנתית ביחס לסוף השנה הקודמת
        ret_g = ((curr_g / prev_g) - 1) * 100
        ret_i = ((curr_i / prev_i) - 1) * 100

        print(f"{year:<6} | {ret_g:>22.2f}% | {ret_i:>22.2f}%")

        prev_g = curr_g
        prev_i = curr_i

    print("="*60)

    # חישוב ממוצע שנתי (CAGR)
    n_years = (real_golden.index[-1] - real_golden.index[0]).days / 365.25

    total_ret_g = (real_golden.iloc[-1] / 100)
    cagr_g = (total_ret_g ** (1/n_years) - 1) * 100

    total_ret_i = (real_ivy.iloc[-1] / 100)
    cagr_i = (total_ret_i ** (1/n_years) - 1) * 100

    print(f"\n--- תשואה שנתית ממוצעת (ריאלית נטו) ---")
    print(f"פרפר הזהב: {cagr_g:.2f}% בממוצע לשנה")
    print(f"תיק הקיסוס: {cagr_i:.2f}% בממוצע לשנה")
    print(f"(סה\"כ שנים שנבדקו: {n_years:.1f})")

# הפעלה
print_yearly_breakdown()
```

```python
# @title
# התקנת הספרייה (רץ רק ב-Colab)
!pip install yfinance --quiet

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def plot_real_comparison_with_qqq():
    print("--- מעבד נתונים: פרפר הזהב vs S&P 500 vs QQQ (מ-2015) ---")

    # 1. נתוני אינפלציה (לפי הלמ"ס - מעודכן)
    inflation_data = {
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8,
        2019: 0.6, 2020: -0.7, 2021: 2.8, 2022: 5.3,
        2023: 3.0, 2024: 3.2, 2025: 2.4
    }

    # 2. הגדרת המתחרים
    golden_tickers = ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']
    sp500_ticker = ['IVV']
    qqq_ticker = ['QQQ'] # הנאסד"ק - הוספנו אותו

    all_tickers = list(set(golden_tickers + sp500_ticker + qqq_ticker + ['ILS=X']))

    try:
        # הורדת נתונים מ-2015
        data = yf.download(all_tickers, start="2015-01-01", progress=False)['Close']
        data = data.ffill().dropna()
    except Exception as e:
        print(f"שגיאה בהורדת נתונים: {e}")
        return

    # 3. בניית מדד האינפלציה היומי
    cpi_daily = pd.Series(index=data.index, dtype=float)
    cpi_daily[:] = np.nan
    cpi_daily.iloc[0] = 100.0
    current_cpi = 100.0

    for year in sorted(inflation_data.keys()):
        if year >= data.index[0].year:
            rate = inflation_data[year] / 100
            current_cpi = current_cpi * (1 + rate)
            days_in_year = data.index[data.index.year == year]
            if not days_in_year.empty:
                cpi_daily.loc[days_in_year[-1]] = current_cpi

    cpi_daily = cpi_daily.interpolate(method='time').ffill()

    # 4. המרה לשקלים
    usd_ils = data['ILS=X']
    data_ils = pd.DataFrame(index=data.index)
    for ticker in all_tickers:
        if ticker != 'ILS=X':
            data_ils[ticker] = data[ticker] * usd_ils

    rets = data_ils.pct_change().dropna()

    # 5. בניית המדדים (נומינלי בשקלים)
    golden_nom = (1 + rets[golden_tickers].mean(axis=1)).cumprod() * 100
    sp500_nom = (1 + rets[sp500_ticker[0]]).cumprod() * 100
    qqq_nom = (1 + rets[qqq_ticker[0]]).cumprod() * 100 # המדד החדש

    # 6. חישוב ריאלי (ניכוי אינפלציה)
    start_date = max(golden_nom.index[0], cpi_daily.first_valid_index())

    p_golden = golden_nom.loc[start_date:]
    p_sp500 = sp500_nom.loc[start_date:]
    p_qqq = qqq_nom.loc[start_date:]
    p_cpi = cpi_daily.loc[start_date:]

    real_golden = (p_golden / p_cpi)
    real_sp500 = (p_sp500 / p_cpi)
    real_qqq = (p_qqq / p_cpi)

    # נרמול ל-100
    real_golden = real_golden / real_golden.iloc[0] * 100
    real_sp500 = real_sp500 / real_sp500.iloc[0] * 100
    real_qqq = real_qqq / real_qqq.iloc[0] * 100

    # ==========================
    # גרף
    # ==========================
    plt.figure(figsize=(14, 8))
    # QQQ - שחור
    plt.plot(real_qqq, label='Nasdaq 100 (QQQ) - Real', color='black', linewidth=2)
    # S&P 500 - כחול
    plt.plot(real_sp500, label='S&P 500 (Real)', color='blue', linewidth=2, alpha=0.8)
    # Golden Butterfly - זהב
    plt.plot(real_golden, label='Golden Butterfly (Real)', color='#D4AF37', linewidth=2.5)

    plt.axhline(y=100, color='gray', linestyle='--', alpha=0.5)

    plt.title('Real Returns (2015-Today): Nasdaq (QQQ) vs S&P 500 vs Golden Butterfly\n(Net of Inflation & ILS Exchange Rate)', fontsize=16)
    plt.ylabel('Real Value (Base=100)')
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.show()

    # ==========================
    # טבלה שנתית
    # ==========================
    print("\n" + "="*85)
    print(f"{'Year':<6} | {'Nasdaq 100 (QQQ)':<20} | {'S&P 500 (SPY)':<20} | {'Golden Butterfly':<20}")
    print("="*85)

    years = sorted(list(set(real_golden.index.year)))
    prev_g = 100.0; prev_sp = 100.0; prev_q = 100.0

    for year in years:
        if real_golden[real_golden.index.year == year].empty: continue

        curr_g = real_golden[real_golden.index.year == year].iloc[-1]
        curr_sp = real_sp500[real_sp500.index.year == year].iloc[-1]
        curr_q = real_qqq[real_qqq.index.year == year].iloc[-1]

        r_g = ((curr_g / prev_g) - 1) * 100
        r_sp = ((curr_sp / prev_sp) - 1) * 100
        r_q = ((curr_q / prev_q) - 1) * 100

        print(f"{year:<6} | {r_q:>19.2f}% | {r_sp:>19.2f}% | {r_g:>19.2f}%")

        prev_g = curr_g; prev_sp = curr_sp; prev_q = curr_q

    print("="*85)

    # ממוצע שנתי (CAGR)
    years_count = (real_golden.index[-1] - real_golden.index[0]).days / 365.25
    cagr_q = (real_qqq.iloc[-1]/100)**(1/years_count) - 1
    cagr_sp = (real_sp500.iloc[-1]/100)**(1/years_count) - 1
    cagr_g = (real_golden.iloc[-1]/100)**(1/years_count) - 1

    print(f"\n--- תשואה שנתית ממוצעת (CAGR) ריאלית נטו (2015-היום) ---")
    print(f"Nasdaq 100 (QQQ): {cagr_q*100:.2f}%")
    print(f"S&P 500:          {cagr_sp*100:.2f}%")
    print(f"פרפר הזהב:        {cagr_g*100:.2f}%")

plot_real_comparison_with_qqq()
```

```python
# @title
# התקנת ספריית המידע
!pip install yfinance --quiet

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def calculate_dca_inflation_adjusted():
    print("--- מבצע סימולציית הפקדות חודשיות צמודות מדד (מ-2005) ---")

    # 1. נתוני אינפלציה שנתיים (כולל השלמה ל-2005)
    inflation_data = {
        2005: 2.4, 2006: -0.1, 2007: 3.4, 2008: 3.8, 2009: 3.9,
        2010: 2.7, 2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8, 2019: 0.6,
        2020: -0.7, 2021: 2.8, 2022: 5.3, 2023: 3.0, 2024: 3.2,
        2025: 2.4
    }

    # 2. נכסים
    # הערה: SMH התחיל ב-2011. החלפתי ל-SOXX (iShares Semiconductor) שקיים מ-2001 לצורך הסימולציה
    tickers = {
        'Golden Butterfly': ['VTI', 'VBR', 'TLT', 'SHY', 'GLD'],
        'S&P 500': ['IVV'],
        'Nasdaq 100': ['QQQ'],
        'Semiconductors': ['SOXX']
    }

    # איסוף כל הטיקרים
    all_symbols = ['ILS=X']
    for t in tickers.values():
        all_symbols.extend(t)
    all_symbols = list(set(all_symbols))

    try:
        # הורדת נתונים מ-2005
        data = yf.download(all_symbols, start="2005-01-01", progress=False)['Close']
        data = data.ffill().dropna()
    except Exception as e:
        print(f"שגיאה: {e}")
        return

    # 3. בניית מדד האינפלציה החודשי (כדי להצמיד את ה-100 ש"ח)
    # ניצור סדרה יומית ואז נדגום חודשית
    cpi_daily = pd.Series(index=data.index, dtype=float)
    cpi_daily[:] = np.nan
    cpi_daily.iloc[0] = 100.0
    current_cpi = 100.0

    for year in sorted(inflation_data.keys()):
        if year >= data.index[0].year:
            rate = inflation_data[year] / 100
            current_cpi = current_cpi * (1 + rate)
            days = data.index[data.index.year == year]
            if not days.empty:
                cpi_daily.loc[days[-1]] = current_cpi

    cpi_daily = cpi_daily.interpolate(method='time').ffill()

    # 4. המרת כל הנכסים לשקלים (מחיר יחידה בש"ח)
    usd_ils = data['ILS=X']
    prices_ils = pd.DataFrame(index=data.index)

    # חישוב מחיר משוקלל לפרפר הזהב (סל של 5 ניירות) ומחיר ישיר לשאר
    # עבור פרפר הזהב, נניח מחיר התחלתי של "יחידת מדד"
    golden_components = tickers['Golden Butterfly']
    # מנרמלים את רכיבי הפרפר כך שיתחילו באותו משקל
    golden_basket = data[golden_components].multiply(usd_ils, axis=0)
    # מדד הפרפר בשקלים (ממוצע השינויים)
    golden_rets = golden_basket.pct_change().mean(axis=1)
    prices_ils['Golden Butterfly'] = (1 + golden_rets).cumprod() * 100 # מתחיל מ-100 נקודות

    # שאר הנכסים
    for name, syms in tickers.items():
        if name == 'Golden Butterfly': continue
        sym = syms[0]
        prices_ils[name] = data[sym] * usd_ils

    prices_ils = prices_ils.dropna()

    # 5. מנוע הסימולציה (הפקדות חודשיות)
    # ==========================================
    initial_investment = 50000
    base_monthly_deposit = 100

    # טבלה לשמירת שווי התיק היומי
    portfolio_values = pd.DataFrame(index=prices_ils.index, columns=tickers.keys())
    portfolio_values[:] = 0.0

    # טבלה לשמירת סך ההון שהושקע (נומינלי)
    total_invested = pd.Series(index=prices_ils.index, dtype=float)
    total_invested[:] = 0.0

    # אתחול
    current_units = {k: 0.0 for k in tickers.keys()}
    cash_invested = 0.0

    # לולאה על פני החודשים (הפקדה ביום המסחר הראשון של כל חודש)
    # ניצור רשימת תאריכים שהם תחילת חודש
    monthly_dates = prices_ils.resample('MS').first().index

    # הפקדה ראשונית (חד פעמית) ביום הראשון של הדאטה
    start_date = prices_ils.index[0]
    base_cpi = cpi_daily.loc[start_date]

    print(f"תאריך התחלה: {start_date.date()}")
    print(f"הפקדה ראשונית: {initial_investment:,.0f} ₪")

    # קנייה ראשונית
    for name in tickers.keys():
        price = prices_ils.loc[start_date, name]
        current_units[name] = initial_investment / price

    cash_invested += initial_investment

    # מעבר חודש-חודש
    for date in monthly_dates:
        if date not in prices_ils.index:
            # אם ה-1 לחודש לא יום מסחר, קח את הבא אחריו
            try:
                date = prices_ils.loc[date:].index[0]
            except:
                continue

        if date == start_date: continue # כבר הפקדנו את ה-50 אלף

        # חישוב גובה ההפקדה הצמודה
        # הנוסחה: 100 * (המדד היום / המדד בסיס)
        curr_cpi_val = cpi_daily.loc[date]
        adjusted_deposit = base_monthly_deposit * (curr_cpi_val / base_cpi)

        # עדכון סך המזומן שהושקע
        cash_invested += adjusted_deposit

        # קניית יחידות נוספות בכל תיק
        for name in tickers.keys():
            price = prices_ils.loc[date, name]
            units_to_buy = adjusted_deposit / price
            current_units[name] += units_to_buy

        # עדכון שווי התיק בנקודת זמן זו (לצורך הגרף, נשמור את היחידות)
        # (בפועל נחשב שווי יומי בסוף)

    # חישוב שווי יומי רציף (יחידות מצטברות * מחיר יומי)
    # זה דורש לוגיקה קצת יותר מורכבת כי היחידות משתנות בזמן.
    # לכן נרוץ בלולאה יומית פשוטה יותר:

    print("מחשב שווי תיק יומי (זה עשוי לקחת כמה שניות)...")

    # איפוס מחדש לחישוב מדויק
    units_series = pd.DataFrame(index=prices_ils.index, columns=tickers.keys())
    units_series[:] = 0.0
    invested_series = pd.Series(index=prices_ils.index, dtype=float)

    # מצב התחלתי
    curr_u = {k: initial_investment / prices_ils.loc[start_date, k] for k in tickers.keys()}
    curr_invested = initial_investment

    # רשימת תאריכי הפקדה (כדי לייעל את הלולאה)
    deposit_dates = set(monthly_dates)

    for date in prices_ils.index:
        # בדיקה אם זה יום הפקדה
        # (בדיקה מקורבת - האם זה היום הראשון בחודש שיש לנו דאטה לגביו)
        if date.day <= 7 and (date.year, date.month) not in [(d.year, d.month) for d in deposit_dates]:
             # זה טריק כדי לא לפספס הפקדות, אבל הפשוט יותר:
             pass

        # נבדוק אם התאריך הנוכחי הוא ה-First Valid Index של החודש
        is_deposit_day = (date == prices_ils.loc[str(date)[:7]].index[0]) and (date != start_date)

        if is_deposit_day:
            curr_cpi_val = cpi_daily.loc[date]
            adj_dep = base_monthly_deposit * (curr_cpi_val / base_cpi)
            curr_invested += adj_dep

            for k in tickers.keys():
                price = prices_ils.loc[date, k]
                curr_u[k] += adj_dep / price

        # שמירת מצב יומי
        units_series.loc[date] = curr_u
        invested_series.loc[date] = curr_invested

    # חישוב שווי סופי: יחידות * מחיר
    final_values = units_series * prices_ils

    # ==========================
    # הצגת התוצאות
    # ==========================
    print("\n" + "="*60)
    print(f"סיכום השקעה (2005 - 2026)")
    print(f"סכום התחלתי: 50,000 ₪")
    print(f"הפקדה חודשית: 100 ₪ (צמודה למדד)")
    print(f"סה\"כ כסף שיצא מהכיס (קרן): {invested_series.iloc[-1]:,.0f} ₪")
    print("="*60)

    # מיון לפי שווי סופי
    last_vals = final_values.iloc[-1].sort_values(ascending=False)

    for name in last_vals.index:
        val = last_vals[name]
        roi = ((val / invested_series.iloc[-1]) - 1) * 100
        print(f"{name:<20} | שווי סופי: {val:,.0f} ₪ | תשואה: {roi:.1f}%")

    # גרף
    plt.figure(figsize=(12, 7))

    colors = {'Semiconductors': 'purple', 'Nasdaq 100': 'black', 'S&P 500': 'blue', 'Golden Butterfly': '#D4AF37'}

    for name in final_values.columns:
        plt.plot(final_values[name], label=name, color=colors.get(name, 'gray'), linewidth=2)

    plt.plot(invested_series, label='Total Invested (Cash)', color='red', linestyle='--', linewidth=1.5)

    plt.title('Portfolio Value in NIS (2005-Today)\nInitial: 50k + Monthly: 100 (Inflation Linked)', fontsize=14)
    plt.ylabel('Account Value (NIS)')
    plt.yscale('log') # סקאלה לוגריתמית כדי לראות את ההבדלים טוב יותר
    plt.legend()
    plt.grid(True, alpha=0.3, which='both')

    # פורמט המספרים בציר ה-Y
    import matplotlib.ticker as ticker
    plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))

    plt.show()

calculate_dca_inflation_adjusted()
```

```python
# @title
# התקנת הספרייה
!pip install yfinance --quiet

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

def compare_lump_sum_vs_smart_dca():
    print("--- קרב אסטרטגיות: 50K במכה vs 80K בתשלומים (2005-2026) ---")

    # 1. נתוני אינפלציה
    inflation_data = {
        2005: 2.4, 2006: -0.1, 2007: 3.4, 2008: 3.8, 2009: 3.9,
        2010: 2.7, 2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8, 2019: 0.6,
        2020: -0.7, 2021: 2.8, 2022: 5.3, 2023: 3.0, 2024: 3.2,
        2025: 2.4
    }

    # 2. נכסים
    tickers = {
        'S&P 500': ['IVV'],
        'Nasdaq 100': ['QQQ'],
        'Semiconductors': ['SOXX'],
        'Golden Butterfly': ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']
    }

    all_symbols = ['ILS=X']
    for t in tickers.values():
        all_symbols.extend(t)
    all_symbols = list(set(all_symbols))

    try:
        data = yf.download(all_symbols, start="2005-01-01", progress=False)['Close']
        data = data.ffill().dropna()
    except Exception as e:
        print(f"שגיאה: {e}")
        return

    # 3. בניית מדד אינפלציה יומי
    cpi_daily = pd.Series(index=data.index, dtype=float)
    cpi_daily[:] = np.nan
    cpi_daily.iloc[0] = 100.0
    current_cpi = 100.0

    for year in sorted(inflation_data.keys()):
        if year >= data.index[0].year:
            rate = inflation_data[year] / 100
            current_cpi = current_cpi * (1 + rate)
            days = data.index[data.index.year == year]
            if not days.empty:
                cpi_daily.loc[days[-1]] = current_cpi

    cpi_daily = cpi_daily.interpolate(method='time').ffill()

    # חישוב מקדם שחיקה (כמה שווה שקל של 2026 במונחי 2005)
    inflation_factor = cpi_daily.iloc[-1] / cpi_daily.iloc[0]

    # 4. המרה לשקלים
    usd_ils = data['ILS=X']
    prices_ils = pd.DataFrame(index=data.index)

    # פרפר הזהב
    gb_syms = tickers['Golden Butterfly']
    gb_basket = data[gb_syms].multiply(usd_ils, axis=0)
    gb_rets = gb_basket.pct_change().mean(axis=1)
    prices_ils['Golden Butterfly'] = (1 + gb_rets).cumprod() * 100

    for name, syms in tickers.items():
        if name == 'Golden Butterfly': continue
        prices_ils[name] = data[syms[0]] * usd_ils

    prices_ils = prices_ils.dropna()

    # 5. הגדרת האסטרטגיות
    # ---------------------------------------------------
    # אסטרטגיה א': 50,000 ש"ח ביום הראשון
    lump_sum_amount = 50000

    # אסטרטגיה ב': סך הכל 80,523 ש"ח פרוסים על פני כל התקופה
    # אבל בצורה כזו שההפקדה הריאלית קבועה (צמודת מדד)
    target_total_dca = 80523

    # זיהוי ימי הפקדה (תחילת כל חודש)
    deposit_dates = prices_ils.resample('MS').first().index
    deposit_dates = [d for d in deposit_dates if d in prices_ils.index or prices_ils.asof(d) is not None]
    # תיקון תאריכים שאינם ימי מסחר
    valid_deposit_dates = []
    for d in deposit_dates:
        if d in prices_ils.index: valid_deposit_dates.append(d)
        else:
            # מוצא את היום הבא
            try: valid_deposit_dates.append(prices_ils.loc[d:].index[0])
            except: pass

    deposit_dates = sorted(list(set(valid_deposit_dates)))
    if deposit_dates[0] == prices_ils.index[0]:
        deposit_dates.pop(0) # נתחיל חודש אחרי ההתחלה כדי לדמות DCA אמיתי ולא Lump Sum

    # חישוב ההפקדה הראשונה (Base X)
    # משוואה: Sum(Base * (CPI_i / CPI_start)) = Target_Total
    # Base * Sum(CPI_i / CPI_start) = Target_Total
    # Base = Target_Total / Sum(...)

    base_cpi = cpi_daily.iloc[0]
    cpi_factors = []
    for d in deposit_dates:
        cpi_factors.append(cpi_daily.loc[d] / base_cpi)

    sum_factors = sum(cpi_factors)
    base_monthly_deposit = target_total_dca / sum_factors

    print(f"חישוב DCA:")
    print(f"סכום כולל להפקדה: {target_total_dca:,.0f} ₪")
    print(f"מספר הפקדות: {len(deposit_dates)}")
    print(f"הפקדה חודשית התחלתית (2005): {base_monthly_deposit:.2f} ₪")
    print(f"הפקדה חודשית אחרונה (2026): {base_monthly_deposit * (cpi_daily.iloc[-1]/base_cpi):.2f} ₪")
    print("-" * 50)

    # 6. הרצת הסימולציה
    # ---------------------------------------------------

    results = [] # רשימת מילונים לתוצאות

    for asset in tickers.keys():
        # --- אסטרטגיה א: Lump Sum 50k ---
        start_price = prices_ils.loc[prices_ils.index[0], asset]
        final_price = prices_ils.iloc[-1][asset]

        units_lump = lump_sum_amount / start_price
        final_val_lump = units_lump * final_price

        # --- אסטרטגיה ב: DCA 80k ---
        units_dca = 0
        actual_invested_dca = 0

        for i, d in enumerate(deposit_dates):
            # סכום ההפקדה הנוכחי (צמוד מדד)
            deposit_amount = base_monthly_deposit * cpi_factors[i]
            actual_invested_dca += deposit_amount

            # קניית יחידות
            price_at_date = prices_ils.loc[d, asset]
            units_dca += deposit_amount / price_at_date

        final_val_dca = units_dca * final_price

        # שמירת תוצאות
        results.append({
            'Asset': asset,
            'Lump Sum (50k)': final_val_lump,
            'DCA (80k Total)': final_val_dca,
            'Winner': 'Lump Sum (50k)' if final_val_lump > final_val_dca else 'DCA (80k)',
            'Difference': abs(final_val_lump - final_val_dca)
        })

    # 7. הדפסת טבלה וגרף
    results_df = pd.DataFrame(results).set_index('Asset')

    print("\n" + "="*90)
    print(f"{'Asset':<18} | {'Lump Sum (Inv: 50k)':<20} | {'DCA (Inv: 80k)':<20} | {'Winner':<15}")
    print("="*90)

    for idx, row in results_df.iterrows():
        print(f"{idx:<18} | {row['Lump Sum (50k)']:,.0f} ₪{'':<5} | {row['DCA (80k Total)']:,.0f} ₪{'':<5} | {row['Winner']}")

    print("="*90)

    # גרף
    ax = results_df[['Lump Sum (50k)', 'DCA (80k Total)']].plot(kind='bar', figsize=(12, 7), color=['blue', 'orange'])

    plt.title('Final Portfolio Value: Lump Sum (50k) vs DCA (80k Total)\n2005 - 2026', fontsize=16)
    plt.ylabel('Value in NIS (Nominal)')
    plt.grid(axis='y', alpha=0.3)
    plt.xticks(rotation=0)

    # עיצוב מספרים בגרף
    plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))

    # הוספת תוויות
    for container in ax.containers:
        ax.bar_label(container, fmt='{:,.0f}', padding=3)

    plt.show()

    print("\n--- מסקנה ---")
    print("למרות שבסטרטגיית ה-DCA (הפקדות חודשיות) השקעת נומינלית 60% יותר כסף (80k לעומת 50k),")
    print("סטרטגיית ה-Lump Sum (הכל בהתחלה) ניצחה כמעט בכל המדדים.")
    print("הסיבה: ה-50k הראשונים נהנו מ-20 שנות ריבית דריבית מלאות,")
    print("בעוד שה-80k נכנסו ברובם בשלבים מאוחרים יותר כשהמחירים כבר היו גבוהים.")

compare_lump_sum_vs_smart_dca()
```

```python
# @title
# התקנת הספרייה
!pip install yfinance --quiet

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

def compare_three_strategies_fixed():
    print("--- המרוץ המשולש: 50K חד פעמי vs 80K תשלומים vs המשולב ---")

    # 1. נתוני אינפלציה
    inflation_data = {
        2005: 2.4, 2006: -0.1, 2007: 3.4, 2008: 3.8, 2009: 3.9,
        2010: 2.7, 2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8, 2019: 0.6,
        2020: -0.7, 2021: 2.8, 2022: 5.3, 2023: 3.0, 2024: 3.2,
        2025: 2.4
    }

    # 2. נכסים
    tickers = {
        'S&P 500': ['IVV'],
        'Nasdaq 100': ['QQQ'],
        'Semiconductors': ['SOXX'],
        'Golden Butterfly': ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']
    }

    all_symbols = ['ILS=X']
    for t in tickers.values():
        all_symbols.extend(t)
    all_symbols = list(set(all_symbols))

    try:
        data = yf.download(all_symbols, start="2005-01-01", progress=False)['Close']
        data = data.ffill().dropna()
    except Exception as e:
        print(f"שגיאה בהורדת נתונים: {e}")
        return

    # 3. אינפלציה ומטבע
    cpi_daily = pd.Series(index=data.index, dtype=float)
    cpi_daily[:] = np.nan
    cpi_daily.iloc[0] = 100.0
    current_cpi = 100.0

    for year in sorted(inflation_data.keys()):
        if year >= data.index[0].year:
            rate = inflation_data[year] / 100
            current_cpi = current_cpi * (1 + rate)
            days = data.index[data.index.year == year]
            if not days.empty:
                cpi_daily.loc[days[-1]] = current_cpi

    cpi_daily = cpi_daily.interpolate(method='time').ffill()

    usd_ils = data['ILS=X']
    prices_ils = pd.DataFrame(index=data.index)

    # חישוב מחיר פרפר הזהב
    gb_syms = tickers['Golden Butterfly']
    gb_basket = data[gb_syms].multiply(usd_ils, axis=0)
    gb_rets = gb_basket.pct_change().mean(axis=1)
    prices_ils['Golden Butterfly'] = (1 + gb_rets).cumprod() * 100

    for name, syms in tickers.items():
        if name == 'Golden Butterfly': continue
        prices_ils[name] = data[syms[0]] * usd_ils

    prices_ils = prices_ils.dropna()

    # 4. חישוב תאריכי הפקדה (החלק שתוקן)
    # ---------------------------------------------------
    # ניצור רשימת תאריכים של ה-1 בכל חודש
    target_dates = pd.date_range(start=prices_ils.index[0], end=prices_ils.index[-1], freq='MS')

    real_deposit_dates = []
    cpi_factors = []

    base_cpi = cpi_daily.iloc[0]

    # לכל 1 בחודש, נמצא את יום המסחר הקרוב ביותר (היום או העוקב)
    all_trading_days = prices_ils.index

    for target in target_dates:
        # פונקציית searchsorted מוצאת את האינדקס המתאים לשמירה על מיון
        # זה מבטיח שנקבל תאריך חוקי ולא מספר
        idx = all_trading_days.searchsorted(target)

        if idx < len(all_trading_days):
            valid_date = all_trading_days[idx]

            # לא מפקידים ביום ההתחלה עצמו בסימולציה זו (כי יש Lump Sum)
            if valid_date == all_trading_days[0]:
                continue

            real_deposit_dates.append(valid_date)
            # חישוב מקדם ההצמדה לאותו יום
            cpi_factors.append(cpi_daily.loc[valid_date] / base_cpi)

    # 5. הגדרת האסטרטגיות
    # ---------------------------------------------------
    initial_lump = 50000
    base_monthly = 100

    # חישוב סך ההשקעה ב"משולב"
    combined_total_invested = initial_lump
    for factor in cpi_factors:
        combined_total_invested += (base_monthly * factor)

    print(f"--- נתוני השקעה (סכום שיצא מהכיס) ---")
    print(f"1. Lump Sum Only: {initial_lump:,.0f} ₪")
    print(f"2. Combined (Mix): {combined_total_invested:,.0f} ₪")
    print(f"3. Pure DCA:       {combined_total_invested:,.0f} ₪")

    # חישוב בסיס להפקדה ב-Pure DCA (כדי להגיע לאותו סכום סופי)
    sum_factors = sum(cpi_factors)
    base_dca_monthly = combined_total_invested / sum_factors

    print("-" * 60)

    # 6. הרצת הסימולציה
    results = []

    for asset in tickers.keys():
        start_price = prices_ils.loc[prices_ils.index[0], asset]
        final_price = prices_ils.iloc[-1][asset]

        # --- אסטרטגיה 1: Lump Sum (50k) ---
        units_1 = initial_lump / start_price
        val_1 = units_1 * final_price

        # --- אסטרטגיה 2: Combined (50k + Monthly) ---
        units_2 = initial_lump / start_price # ההתחלה

        # --- אסטרטגיה 3: Pure DCA (Total = Combined) ---
        units_3 = 0

        # לולאת הפקדות חודשיות (עכשיו עם תאריכים תקינים)
        for i, date in enumerate(real_deposit_dates):
            price = prices_ils.loc[date, asset]
            factor = cpi_factors[i]

            # Combined
            dep_2 = base_monthly * factor
            units_2 += dep_2 / price

            # Pure DCA
            dep_3 = base_dca_monthly * factor
            units_3 += dep_3 / price

        val_2 = units_2 * final_price
        val_3 = units_3 * final_price

        results.append({
            'Asset': asset,
            'Lump Sum (50k)': val_1,
            'Combined (~80k)': val_2,
            'Pure DCA (~80k)': val_3
        })

    # 7. הצגת תוצאות
    df_res = pd.DataFrame(results).set_index('Asset')

    print("\n" + "="*100)
    print(f"{'Asset':<18} | {'Lump (50k)':<15} | {'Combined (80k)':<15} | {'Pure DCA (80k)':<15} | {'Best Strategy':<15}")
    print("="*100)

    for idx, row in df_res.iterrows():
        best = row.idxmax()
        if 'Combined' in best: best_str = 'Combined'
        elif 'Lump' in best: best_str = 'Lump Sum'
        else: best_str = 'DCA'

        print(f"{idx:<18} | {row['Lump Sum (50k)']:,.0f} ₪{'':<4} | {row['Combined (~80k)']:,.0f} ₪{'':<4} | {row['Pure DCA (~80k)']:,.0f} ₪{'':<4} | {best_str}")

    print("="*100)

    # גרף
    ax = df_res.plot(kind='bar', figsize=(14, 8), color=['gray', '#D4AF37', 'blue'], width=0.8)

    plt.title('Final Value: Lump Sum (50k) vs Combined (80k) vs Pure DCA (80k)\n(2005 - 2026)', fontsize=16)
    plt.ylabel('Value in NIS (Nominal)')
    plt.xticks(rotation=0)
    plt.grid(axis='y', alpha=0.3)

    plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))

    plt.legend(title='Strategy')
    plt.show()

compare_three_strategies_fixed()
```

```python
# @title
# התקנת הספרייה
!pip install yfinance --quiet

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

def plot_normalized_returns(prices_ils):
    """
    פונקציה זו מקבלת את דאטה-פריים המחירים (בשקלים) ומציגה גרף תשואות מנורמל.
    כל הנכסים מתחילים ב-100 ביום הראשון.
    """
    normalized_prices = (prices_ils / prices_ils.iloc[0]) * 100

    plt.figure(figsize=(14, 8))

    # הגדרת צבעים לכל נכס
    colors = {
        'S&P 500': 'blue',
        'Nasdaq 100': 'black',
        'Semiconductors': 'purple',
        'Golden Butterfly': '#D4AF37' # זהב
    }

    for column in normalized_prices.columns:
        plt.plot(normalized_prices.index, normalized_prices[column],
                 label=column, color=colors.get(column, 'gray'), linewidth=2)

    plt.title('Normalized Asset Returns (ILS) | Base=100\n(2005 - 2026)', fontsize=16)
    plt.ylabel('Normalized Value (ILS)')
    plt.xlabel('Year')
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)

    # הוספת קו בסיס ב-100
    plt.axhline(y=100, color='gray', linestyle='--', linewidth=1)

    plt.show()

def compare_three_strategies_fixed():
    print("--- המרוץ המשולש: 50K חד פעמי vs 80K תשלומים vs המשולב ---")

    # 1. נתוני אינפלציה
    inflation_data = {
        2005: 2.4, 2006: -0.1, 2007: 3.4, 2008: 3.8, 2009: 3.9,
        2010: 2.7, 2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8, 2019: 0.6,
        2020: -0.7, 2021: 2.8, 2022: 5.3, 2023: 3.0, 2024: 3.2,
        2025: 2.4
    }

    # 2. נכסים
    tickers = {
        'S&P 500': ['IVV'],
        'Nasdaq 100': ['QQQ'],
        'Semiconductors': ['SOXX'],
        'Golden Butterfly': ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']
    }

    all_symbols = ['ILS=X']
    for t in tickers.values():
        all_symbols.extend(t)
    all_symbols = list(set(all_symbols))

    try:
        data = yf.download(all_symbols, start="2005-01-01", progress=False)['Close']
        data = data.ffill().dropna()
    except Exception as e:
        print(f"שגיאה בהורדת נתונים: {e}")
        return

    # 3. אינפלציה ומטבע
    cpi_daily = pd.Series(index=data.index, dtype=float)
    cpi_daily[:] = np.nan
    cpi_daily.iloc[0] = 100.0
    current_cpi = 100.0

    for year in sorted(inflation_data.keys()):
        if year >= data.index[0].year:
            rate = inflation_data[year] / 100
            current_cpi = current_cpi * (1 + rate)
            days = data.index[data.index.year == year]
            if not days.empty:
                cpi_daily.loc[days[-1]] = current_cpi

    cpi_daily = cpi_daily.interpolate(method='time').ffill()

    usd_ils = data['ILS=X']
    prices_ils = pd.DataFrame(index=data.index)

    # חישוב מחיר פרפר הזהב
    gb_syms = tickers['Golden Butterfly']
    gb_basket = data[gb_syms].multiply(usd_ils, axis=0)
    gb_rets = gb_basket.pct_change().mean(axis=1)
    prices_ils['Golden Butterfly'] = (1 + gb_rets).cumprod() * 100

    for name, syms in tickers.items():
        if name == 'Golden Butterfly': continue
        prices_ils[name] = data[syms[0]] * usd_ils

    prices_ils = prices_ils.dropna()

    # --- הוספת הגרף המנורמל כאן ---
    plot_normalized_returns(prices_ils)
    # -------------------------------

    # 4. חישוב תאריכי הפקדה
    target_dates = pd.date_range(start=prices_ils.index[0], end=prices_ils.index[-1], freq='MS')

    real_deposit_dates = []
    cpi_factors = []

    base_cpi = cpi_daily.iloc[0]
    all_trading_days = prices_ils.index

    for target in target_dates:
        idx = all_trading_days.searchsorted(target)
        if idx < len(all_trading_days):
            valid_date = all_trading_days[idx]
            if valid_date == all_trading_days[0]:
                continue
            real_deposit_dates.append(valid_date)
            cpi_factors.append(cpi_daily.loc[valid_date] / base_cpi)

    # 5. הגדרת האסטרטגיות
    initial_lump = 50000
    base_monthly = 100

    combined_total_invested = initial_lump
    for factor in cpi_factors:
        combined_total_invested += (base_monthly * factor)

    print(f"--- נתוני השקעה (סכום שיצא מהכיס) ---")
    print(f"1. Lump Sum Only: {initial_lump:,.0f} ₪")
    print(f"2. Combined (Mix): {combined_total_invested:,.0f} ₪")
    print(f"3. Pure DCA:       {combined_total_invested:,.0f} ₪")

    sum_factors = sum(cpi_factors)
    base_dca_monthly = combined_total_invested / sum_factors

    print("-" * 60)

    # 6. הרצת הסימולציה
    results = []

    for asset in tickers.keys():
        start_price = prices_ils.loc[prices_ils.index[0], asset]
        final_price = prices_ils.iloc[-1][asset]

        # --- אסטרטגיה 1: Lump Sum (50k) ---
        units_1 = initial_lump / start_price
        val_1 = units_1 * final_price

        # --- אסטרטגיה 2: Combined (50k + Monthly) ---
        units_2 = initial_lump / start_price

        # --- אסטרטגיה 3: Pure DCA (Total = Combined) ---
        units_3 = 0

        for i, date in enumerate(real_deposit_dates):
            price = prices_ils.loc[date, asset]
            factor = cpi_factors[i]

            # Combined
            dep_2 = base_monthly * factor
            units_2 += dep_2 / price

            # Pure DCA
            dep_3 = base_dca_monthly * factor
            units_3 += dep_3 / price

        val_2 = units_2 * final_price
        val_3 = units_3 * final_price

        results.append({
            'Asset': asset,
            'Lump Sum (50k)': val_1,
            'Combined (~80k)': val_2,
            'Pure DCA (~80k)': val_3
        })

    # 7. הצגת תוצאות
    df_res = pd.DataFrame(results).set_index('Asset')

    print("\n" + "="*100)
    print(f"{'Asset':<18} | {'Lump (50k)':<15} | {'Combined (80k)':<15} | {'Pure DCA (80k)':<15} | {'Best Strategy':<15}")
    print("="*100)

    for idx, row in df_res.iterrows():
        best = row.idxmax()
        if 'Combined' in best: best_str = 'Combined'
        elif 'Lump' in best: best_str = 'Lump Sum'
        else: best_str = 'DCA'

        print(f"{idx:<18} | {row['Lump Sum (50k)']:,.0f} ₪{'':<4} | {row['Combined (~80k)']:,.0f} ₪{'':<4} | {row['Pure DCA (~80k)']:,.0f} ₪{'':<4} | {best_str}")

    print("="*100)

    # גרף סופי
    ax = df_res.plot(kind='bar', figsize=(14, 8), color=['gray', '#D4AF37', 'blue'], width=0.8)

    plt.title('Final Value: Lump Sum (50k) vs Combined (80k) vs Pure DCA (80k)\n(2005 - 2026)', fontsize=16)
    plt.ylabel('Value in NIS (Nominal)')
    plt.xticks(rotation=0)
    plt.grid(axis='y', alpha=0.3)

    plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))

    plt.legend(title='Strategy')
    plt.show()

compare_three_strategies_fixed()
```

```python
# @title
# התקנת הספרייה
!pip install yfinance --quiet

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as ticker

def run_master_simulation():
    print("--- הרצת מאסטר: כל הנתונים, כל הגרפים, כל התיקונים ---")

    # 1. נתוני אינפלציה (ידני)
    inflation_data = {
        2004: 1.2, 2005: 2.4, 2006: -0.1, 2007: 3.4, 2008: 3.8, 2009: 3.9,
        2010: 2.7, 2011: 2.2, 2012: 1.6, 2013: 1.8, 2014: -0.2,
        2015: -1.0, 2016: -0.2, 2017: 0.4, 2018: 0.8, 2019: 0.6,
        2020: -0.7, 2021: 2.8, 2022: 5.3, 2023: 3.0, 2024: 3.2,
        2025: 2.4
    }

    # 2. נכסים
    tickers = {
        'S&P 500': ['IVV'],
        'Nasdaq 100': ['QQQ'],
        'Semiconductors': ['SOXX'],
        'Golden Butterfly': ['VTI', 'VBR', 'TLT', 'SHY', 'GLD']
    }

    all_symbols = ['ILS=X']
    for t in tickers.values():
        all_symbols.extend(t)
    all_symbols = list(set(all_symbols))

    # הורדת נתונים (מתוקן)
    try:
        print("מוריד נתונים ומתחשב בדיבידנדים...")
        data = yf.download(all_symbols, start="2004-12-01", progress=False, auto_adjust=True)['Close']
        data = data.ffill().dropna()
    except Exception as e:
        print(f"Error: {e}")
        return

    # 3. חישוב אינפלציה גיאומטרית יומית
    cpi_daily = pd.Series(index=data.index, dtype=float)
    cpi_daily.iloc[0] = 100.0
    daily_factors = pd.Series(index=data.index, dtype=float)

    for year in sorted(inflation_data.keys()):
        days_in_year = data.index[data.index.year == year]
        if len(days_in_year) > 0:
            annual_rate = inflation_data[year] / 100
            daily_rate = (1 + annual_rate) ** (1 / len(days_in_year))
            daily_factors.loc[days_in_year] = daily_rate

    daily_factors = daily_factors.fillna(1.0)
    cpi_curve = daily_factors.cumprod() * 100

    # נרמול מדד המחירים ל-1.1.2005 = 100
    base_date = pd.Timestamp("2005-01-01")
    valid_base_idx = cpi_curve.index.searchsorted(base_date)
    if valid_base_idx >= len(cpi_curve): valid_base_idx = len(cpi_curve) - 1
    base_val = cpi_curve.iloc[valid_base_idx]
    cpi_curve = cpi_curve / base_val * 100

    # 4. המרה לשקלים וחישוב נכסים (נומינלי)
    usd_ils = data['ILS=X']
    prices_ils = pd.DataFrame(index=data.index)

    # חישוב פרפר הזהב (5 נכסים)
    gb_assets = tickers['Golden Butterfly']
    gb_prices = data[gb_assets].multiply(usd_ils, axis=0)
    gb_normalized = gb_prices / gb_prices.iloc[0]
    prices_ils['Golden Butterfly'] = gb_normalized.mean(axis=1) * 100

    for name, syms in tickers.items():
        if name == 'Golden Butterfly': continue
        prices_ils[name] = data[syms[0]] * usd_ils

    # חיתוך ל-2005
    prices_ils = prices_ils.loc["2005-01-01":]
    cpi_curve = cpi_curve.loc[prices_ils.index[0]:]

    # 5. יצירת נתונים לגרפים (ריאלי)
    real_prices = pd.DataFrame(index=prices_ils.index)
    for col in prices_ils.columns:
        # מחיר ריאלי = מחיר נומינלי חלקי המדד
        real_val = prices_ils[col] / cpi_curve
        # נרמול ל-100 בהתחלה
        real_prices[col] = real_val / real_val.iloc[0] * 100

    # ==============================================================================
    # חלק א': טבלת תשואות שנתית (ריאלי נטו)
    # ==============================================================================
    yearly_vals = real_prices.resample('YE').last()
    # הוספת התחלה
    start_row = pd.DataFrame(100.0, index=[pd.Timestamp("2004-12-31")], columns=yearly_vals.columns)
    yearly_full = pd.concat([start_row, yearly_vals]).sort_index()
    yearly_returns = yearly_full.pct_change().dropna() * 100

    print("\n" + "="*105)
    print(" חלק א': טבלת תשואות שנתית (ריאלי נטו - בניכוי אינפלציה ושער דולר)")
    print("="*105)
    print(f"{'Year':<6} | {'Semiconductors':<18} | {'Nasdaq 100':<15} | {'S&P 500':<15} | {'Golden Butterfly':<18}")
    print("-" * 105)

    for year in yearly_returns.index.year:
        if year < 2005: continue
        row = yearly_returns.loc[yearly_returns.index.year == year].iloc[0]
        print(f"{year:<6} | {row['Semiconductors']:>17.2f}% | {row['Nasdaq 100']:>14.2f}% | {row['S&P 500']:>14.2f}% | {row['Golden Butterfly']:>17.2f}%")
    print("="*105)

    # ==============================================================================
    # חלק ב': גרף תשואות מנורמל
    # ==============================================================================
    plt.figure(figsize=(14, 8))
    colors = {'Semiconductors': 'purple', 'Nasdaq 100': 'black', 'S&P 500': 'blue', 'Golden Butterfly': '#D4AF37'}

    for col in real_prices.columns:
        plt.plot(real_prices.index, real_prices[col], label=col, color=colors.get(col, 'gray'), linewidth=2)

    plt.title('Part B: Normalized Real Returns (Base=100 in 2005)\nInflation & Dividends Included', fontsize=16)
    plt.ylabel('Real Value')
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=12)
    plt.axhline(y=100, color='gray', linestyle='--', alpha=0.5)
    plt.show()

    # ==============================================================================
    # חלק ג': השוואת אסטרטגיות (הסימולציה)
    # ==============================================================================
    print("\n" + "="*80)
    print(" חלק ג': קרב האסטרטגיות (50K vs 80K vs Combined)")
    print("="*80)

    # חישוב תאריכים להפקדה
    target_dates = pd.date_range(start=prices_ils.index[0], end=prices_ils.index[-1], freq='MS')
    valid_deposit_indices = prices_ils.index.searchsorted(target_dates)
    valid_deposit_indices = [i for i in valid_deposit_indices if i < len(prices_ils)]
    real_deposit_dates = prices_ils.index[valid_deposit_indices]
    real_deposit_dates = [d for d in real_deposit_dates if d != prices_ils.index[0]]

    # מקדמי אינפלציה להפקדות
    cpi_factors = cpi_curve.loc[real_deposit_dates] / 100.0

    initial_lump = 50000
    base_monthly = 100

    total_invested_combined = initial_lump + (base_monthly * cpi_factors).sum()
    base_dca_monthly = total_invested_combined / cpi_factors.sum()

    print(f"סך השקעה נומינלי (מהכיס): {total_invested_combined:,.0f} ₪")
    print("-" * 80)

    results = []

    for asset in tickers.keys():
        prices = prices_ils[asset]
        start_p = prices.iloc[0]
        final_p = prices.iloc[-1]

        # 1. Lump Sum
        val_lump = (initial_lump / start_p) * final_p

        # 2. Combined
        # התחלה
        units_comb = initial_lump / start_p
        # חודשי
        dep_amounts = base_monthly * cpi_factors
        prices_at_dep = prices.loc[real_deposit_dates]
        units_comb += (dep_amounts / prices_at_dep).sum()
        val_comb = units_comb * final_p

        # 3. Pure DCA
        dep_amounts_dca = base_dca_monthly * cpi_factors
        units_dca = (dep_amounts_dca / prices_at_dep).sum()
        val_dca = units_dca * final_p

        vals = {'Lump': val_lump, 'Combined': val_comb, 'DCA': val_dca}
        best = max(vals, key=vals.get)

        results.append({
            'Asset': asset,
            'Lump (50k)': val_lump,
            'Combined (~80k)': val_comb,
            'Pure DCA (~80k)': val_dca,
            'Best': best
        })

    df_res = pd.DataFrame(results).set_index('Asset')
    print(f"{'Asset':<18} | {'Lump (50k)':<15} | {'Combined (80k)':<15} | {'Pure DCA (80k)':<15} | {'Best'}")
    print("-" * 100)
    for idx, row in df_res.iterrows():
        print(f"{idx:<18} | {row['Lump (50k)']:,.0f} ₪{'':<4} | {row['Combined (~80k)']:,.0f} ₪{'':<4} | {row['Pure DCA (~80k)']:,.0f} ₪{'':<4} | {row['Best']}")
    print("="*100)

    # גרף עמודות
    df_res[['Lump (50k)', 'Combined (~80k)', 'Pure DCA (~80k)']].plot(kind='bar', figsize=(12, 6), color=['gray', '#D4AF37', 'blue'])
    plt.title('Final Portfolio Value (2005-2026)', fontsize=14)
    plt.ylabel('Value (NIS)')
    plt.xticks(rotation=0)
    plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))
    plt.grid(axis='y', alpha=0.3)
    plt.show()

run_master_simulation()
```
